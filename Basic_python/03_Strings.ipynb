## 1. Intuition and Motivation

### What Is a String, Really?

Every program that communicates with a human -- displaying a message, reading a name, showing an error, formatting a report -- works with text. In Python, text is represented as a **string**: a sequence of characters enclosed in quotes.

A string is more than just text you can print. It is a **structured sequence**  Every character has a position, every position can be accessed, and Python provides a rich set of tools to analyze, transform, and build strings.

---

### The Real-World Analogy

Think of a string like a **train**. Each carriage is a character. The carriages are numbered from the front (position 0, 1, 2...) and also from the back (-1, -2, -3...). You can:

- Access a single carriage by its number (indexing)
- Take a section of carriages from position A to B (slicing)
- Run operations on the whole train (methods)
- Paint the train a new color -- but this creates a *new train*, the original is unchanged (immutability)

---

A beginner can print strings. A confident programmer can:
- Extract exactly the part they need from any text

## 2. Strings as Sequences of Characters

### The Fundamental Idea

A string is not just a blob of text -- it is an **ordered sequence of characters**. Python stores each character at a specific numbered position called an **index**.

Positions start at **0**, not 1. This is called **zero-based indexing**.

```
String:   P  y  t  h  o  n
Index:    0  1  2  3  4  5
Negative:-6 -5 -4 -3 -2 -1
```

Every character has two valid index numbers: a positive one from the left, and a negative one from the right.

````python
# Demonstrating that a string is a sequence
word = "Python"
print("The string:", word)
print("Length:", len(word))
print("Type:", type(word))
````

````python
# Visualizing every index position
word = "Python"
print("Positive indexes:")
for i in range(len(word)):
    print("  Index", i, "->", word[i])
print()
print("Negative indexes:")
for i in range(-len(word), 0):
    print("  Index", i, "->", word[i])
````

## 3. String Indexing

### What Is Indexing?

Indexing means **accessing a single character** from a string by specifying its position in square brackets.

### Syntax

```python
string[index]
```

| Part | Meaning |
|---|---|
| `string` | The string variable |
| `[index]` | The position number in square brackets |
| Positive index | Counts from the left, starting at 0 |
| Negative index | Counts from the right, starting at -1 |

### Why Negative Indexes?

Negative indexing lets you access characters from the **end** without knowing the length. `string[-1]` always gives the last character regardless of how long the string is.

````python
# Basic indexing
name = "Alice"
print("First character  (index  0):", name[0])
print("Second character (index  1):", name[1])
print("Last character   (index  4):", name[4])
print("Last character   (index -1):", name[-1])
print("Second to last   (index -2):", name[-2])
print("First character  (index -5):", name[-5])
````

````python
# Real-world: checking characters in a filename
filename = "report_2024.pdf"
print("Filename:", filename)
print("First character:", filename[0])
print("Last character:", filename[-1])
print("Last 3 chars (extension):", filename[-3:])
````

````python
# Edge case: single-character string
single = "X"
print("Index 0 and -1 on single char:", single[0], "==", single[-1])
print("They are the same:", single[0] == single[-1])
````

````python
# Out-of-range index causes IndexError
word = "Hello"
print("Length:", len(word), "| Valid indexes: 0 to", len(word)-1, "and", -len(word), "to -1")

# Safe access pattern
index = 10
if 0 <= index < len(word):
    print(word[index])
else:
    print("Index", index, "is out of range for string of length", len(word))
````

## 4. String Slicing

### What Is Slicing?

Slicing extracts a **substring** -- a portion of characters from a string. Instead of getting one character, you get a range of them.

### Syntax

```python
string[start : stop : step]
```

| Part | Meaning | Default |
|---|---|---|
| `start` | Index where the slice begins (inclusive) | `0` |
| `stop` | Index where the slice ends (**exclusive** -- not included) | `len(string)` |
| `step` | How many positions to jump each time | `1` |

### The Critical Rule

`stop` is **exclusive**. `string[1:4]` gives characters at positions 1, 2, and 3. Position 4 is not included. Think of it as *up to but not including* stop.

Any part can be omitted: `string[:4]`, `string[2:]`, `string[:]`, `string[::2]`

````python
# Basic slicing
text = "Hello, World!"
print("Full string:          ", text)
print("Characters 0-4 [0:5]:", text[0:5])
print("Characters 7-11[7:12]:", text[7:12])
print("Start to 5   [:5]:   ", text[:5])
print("7 to end     [7:]:   ", text[7:])
print("Full copy    [:]:    ", text[:])
````

````python
# Slicing with step
alphabet = "abcdefghij"
print("Every char   (step 1):", alphabet[::1])
print("Every 2nd    (step 2):", alphabet[::2])
print("Every 3rd    (step 3):", alphabet[::3])
print("Reversed     (step-1):", alphabet[::-1])
print("Rev every 2nd(step-2):", alphabet[::-2])
````

````python
# Reversing a string with [::-1]
word = "Python"
print("Original:", word)
print("Reversed:", word[::-1])

# Palindrome checker
test_words = ["racecar", "level", "hello", "madam", "world"]
for w in test_words:
    label = "palindrome" if w == w[::-1] else "NOT palindrome"
    print(w, "->", label)
````

````python
# Real-world: extracting parts of a date string
date_string = "2024-08-15"
year  = date_string[0:4]
month = date_string[5:7]
day   = date_string[8:10]
print("Full date:", date_string)
print("Year:", year, "| Month:", month, "| Day:", day)
````

````python
# Slicing with negative indexes
filename = "document_final.pdf"
print("Extension (last 3):", filename[-3:])
print("Name without ext:  ", filename[:-4])

# Checking file extension
if filename[-4:] == ".pdf":
    print("This is a PDF file.")
else:
    print("Not a PDF file.")
````

````python
# Edge case: out-of-range slices do NOT raise IndexError
word = "Hi"
print("word[0:100]:", word[0:100])  # Returns as much as exists
print("word[5:10]: ", word[5:10])   # Returns empty string
print("word[3:1]:  ", word[3:1])    # Returns empty string (start > stop)
print("Slicing never crashes -- indexing with word[10] WOULD crash.")
````

## 5. Escape Characters

### What Are Escape Characters?

Some characters cannot be typed directly inside a string:
- A newline would break the string across lines
- A tab is invisible and ambiguous
- A quote inside a quoted string would end the string prematurely

**Escape characters** use a backslash `\` followed by a letter to represent these.

### Common Escape Characters

| Escape | Meaning |
|---|---|
| `\n` | Newline -- move to next line |
| `\t` | Tab -- horizontal whitespace |
| `\\` | Literal backslash character |
| `\"` | Double quote inside a double-quoted string |
| `\'` | Single quote inside a single-quoted string |
| `\r` | Carriage return |

````python
# Basic escape characters
print("Line one\nLine two\nLine three")
print()
print("Name:\tAlice")
print("Score:\t92")
````

````python
# Including quotes inside strings
# Method 1: use backslash to escape
print("She said: \"Hello!\"")
print('He replied: \'Goodbye!\'')
print()
# Method 2 (cleaner): mix quote styles
print("It is a beautiful day.")   # single quote style is fine too
print("He said Python is great.")  # no internal quotes needed here
````

````python
# Backslash itself needs escaping
print("Windows path: C:\\Users\\Alice\\Documents")
print("A single backslash: \\")
print()
# Raw strings: prefix r"..." treats every backslash as a literal character
# Useful for file paths and regular expressions
path = r"C:\Users\Alice\Documents"
print("Raw string path:", path)
````

````python
# Building formatted output with escape characters
print("Student Report")
print("-" * 30)
print("Name:\t\tAlice Johnson")
print("Score:\t\t94")
print("Grade:\t\tA")
print("Comment:\t\"Excellent work.\"")
print("-" * 30)
````

````python
# repr() reveals escape characters inside a string
text = "Hello\nWorld"
print("When printed:")
print(text)
print()
print("repr() shows the raw escape sequences:")
print(repr(text))
print()
print("Length (\\n counts as ONE character):", len(text))
````

## 6. Multiline Strings

### What Are They?

A **multiline string** is a string that literally spans multiple lines in your code. You create one using **triple quotes** -- either `"""` or `'''`.

Everything between the opening and closing triple quotes is part of the string, including actual newlines you press Enter for.

### When to Use Them

- Writing long text blocks (messages, instructions, paragraphs)
- Documentation strings inside functions and classes
- Templates where line breaks are meaningful
- Avoiding repeated `\n` characters in long strings

````python
message = """This is line one.
This is line two.
This is line three."""
print(message)
````

````python
poem = '''Roses are red,
Violets are blue,
Python is powerful,
And so are you.'''
print(poem)
print("Length:", len(poem))
````

````python
# Multiline vs escaped newlines -- both produce the same string
using_escape = "Line one.\nLine two.\nLine three."
using_triple = """Line one.
Line two.
Line three."""
print("Are they equal?", using_escape == using_triple)
print()
print(using_escape)
print()
print(using_triple)
````

````python
# Real-world: an email template stored as a multiline string
email_body = """Dear Alice,

I am writing to update you on the project status.
Everything is on track and we expect completion by Friday.

Please let me know if you have any questions.

Best regards,
The Team"""
print(email_body)
````

````python
# Multiline strings preserve internal quotes without escaping
sql_query = """
SELECT name, age, email
FROM users
WHERE age > 18
ORDER BY name;
"""
print(sql_query)
````

## 7. String Immutability

### What Does Immutable Mean?

In Python, strings are **immutable** -- once created, a string's characters cannot be changed. You cannot modify a character at a specific position, add to the middle, or remove a character in place.

Any operation that appears to modify a string actually **creates a brand new string**.

### Why Are Strings Immutable?

- **Safety** -- passing a string to a function guarantees it cannot be modified
- **Performance** -- Python can cache and reuse identical strings
- **Hashability** -- immutable objects can be used as dictionary keys

### The Real-World Analogy

Think of a string like a **printed book**. You cannot go into the book and change a word. If you want a corrected version, you print a **new book**. The original is unchanged.

````python
# Strings cannot be modified in place
word = "Hello"
# word[0] = "J"   # Uncommenting this causes TypeError
print("word =", word)
print("You cannot change word[0] directly.")
print("You must create a new string instead.")
````

````python
# Correct approach: build a new string
original = "Hello"
modified = "J" + original[1:]   # "J" + "ello"
print("Original:", original)
print("Modified:", modified)
print("Original unchanged:", original)
````

````python
# All string methods return NEW strings -- they never change the original
text = "hello"
upper_text = text.upper()   # Returns a brand new string
print("Original:", text)           # Still lowercase
print("New string:", upper_text)   # New uppercase string
print("Original is still:", text)
````

````python
# Reassigning a variable vs modifying a string
name = "Alice"
print("Before:", name, "| id:", id(name))
name = "Bob"   # The VARIABLE is reassigned -- the string "Alice" is unchanged
print("After: ", name, "| id:", id(name))
print()
print("The variable now points to a different string.")
print("Reassignment is NOT mutation.")
````

````python
# Practical implication: += creates a new string each time
result = ""
result += "Hello"
result += ", "
result += "World"
result += "!"
print(result)
print("Each += created a new string object.")
print("For large-scale string building, use join() instead.")
````

## 8. String Methods

### What Are String Methods?

A **method** is a function that belongs to a specific type. String methods are functions built into every string that let you transform, analyze, and work with text.

Call them with dot notation: `string.method_name(arguments)`

**Because strings are immutable, all methods return a new string -- they never modify the original.**

### Categories of String Methods

| Category | Methods |
|---|---|
| Case conversion | `upper()`, `lower()`, `title()`, `capitalize()`, `swapcase()` |
| Searching | `find()`, `index()`, `count()`, `startswith()`, `endswith()` |
| Cleaning | `strip()`, `lstrip()`, `rstrip()` |
| Splitting / joining | `split()`, `join()`, `splitlines()` |
| Checking | `isdigit()`, `isalpha()`, `isalnum()`, `isspace()`, `isupper()`, `islower()` |
| Replacing | `replace()` |
| Alignment | `center()`, `ljust()`, `rjust()`, `zfill()` |

### 8.1 Case Conversion Methods

````python
text = "hello WORLD from Python"
print("Original:    ", text)
print("upper():     ", text.upper())
print("lower():     ", text.lower())
print("title():     ", text.title())
print("capitalize():", text.capitalize())
print("swapcase():  ", text.swapcase())
````

````python
# Real-world: normalizing user input for case-insensitive comparison
user_input = "   PYTHON   "
if user_input.strip().lower() == "python":
    print("User typed Python (case-insensitive match)")

name = "alice SMITH"
print("Formatted name:", name.title())
````

### 8.2 Searching Methods

````python
# find() returns index of first occurrence, or -1 if not found
sentence = "Python is great. Python is powerful."
print("find(Python):      ", sentence.find("Python"))
print("find(Python, 10):  ", sentence.find("Python", 10))  # search from index 10
print("find(Java):        ", sentence.find("Java"))        # -1 means not found
````

````python
# count() -- how many times a substring appears
text = "banana"
print("count(a):", text.count("a"))
print("count(an):", text.count("an"))
print("count(z):", text.count("z"))  # 0 if not found

sentence = "to be or not to be"
print("count(be) in sentence:", sentence.count("be"))
````

````python
# startswith() and endswith()
filename = "report_2024.pdf"
url = "https://www.example.com"

print("starts with report:", filename.startswith("report"))
print("ends with .pdf:    ", filename.endswith(".pdf"))
print("ends with .txt:    ", filename.endswith(".txt"))

# Check multiple options using a tuple
print("url starts with http or https:", url.startswith(("http://", "https://")))
````

### 8.3 Cleaning Methods

````python
text = "   hello world   "
print("Original: |" + text + "|")
print("strip():  |" + text.strip() + "|")
print("lstrip(): |" + text.lstrip() + "|")
print("rstrip(): |" + text.rstrip() + "|")

# strip() can remove specific characters
messy = "###hello###"
print("strip(#): |" + messy.strip("#") + "|")
````

````python
# Real-world: cleaning user input
raw_inputs = ["  Alice ", "BOB", "  carol  ", "DAVID  "]
for name in raw_inputs:
    cleaned = name.strip().title()
    print("Raw:", repr(name).ljust(14), "-> Cleaned:", repr(cleaned))
````

### 8.4 Splitting and Joining

`split()` breaks one string into a list of strings. `join()` combines a list of strings into one. They are inverses of each other.

````python
# split() -- break a string into a list
sentence = "Python is a powerful language"
words = sentence.split()       # split on whitespace by default
print("split():", words)

csv_line = "Alice,28,Engineer,London"
fields = csv_line.split(",")   # split on comma
print("split(,):", fields)
````

````python
# join() -- combine a list into one string
words = ["Python", "is", "amazing"]
print("space join: ", " ".join(words))
print("dash join:  ", "-".join(words))
print("no join:    ", "".join(words))
print("comma join: ", ", ".join(words))
````

````python
# Real-world: normalize whitespace using split + join
sentence = "  extra   spaces   between   words  "
words = sentence.split()         # split() discards extra whitespace
print("After split():", words)
cleaned = " ".join(words)
print("After join(): ", cleaned)
````

### 8.5 The `replace()` Method

````python
text = "I love cats. Cats are wonderful. My cat is named Luna."
print("Original:          ", text)
print("replace cat->dog:  ", text.replace("cat", "dog"))
print("replace Cat->Dog:  ", text.replace("Cat", "Dog"))  # case-sensitive

# Limit number of replacements
repeated = "aababab"
print("replace a->X (limit 2):", repeated.replace("a", "X", 2))
````

### 8.6 Checking Methods (is...)

````python
# is... methods return True or False
print("String         isalpha  isdigit  isalnum  isspace  isupper  islower")
print("-" * 68)
samples = ["hello", "HELLO", "Hello123", "12345", "   "]
for s in samples:
    row = repr(s).ljust(15)
    row += str(s.isalpha()).ljust(9)
    row += str(s.isdigit()).ljust(9)
    row += str(s.isalnum()).ljust(9)
    row += str(s.isspace()).ljust(9)
    row += str(s.isupper()).ljust(9)
    row += str(s.islower())
    print(row)
````

````python
# Real-world: input validation
def validate_username(username):
    if len(username) < 3:
        return "Too short (minimum 3 characters)"
    if not username.isalnum():
        return "Only letters and numbers allowed"
    if username.isdigit():
        return "Must contain at least one letter"
    return "Valid username"

test_names = ["ab", "alice99", "123456", "alice smith", "Bob2024"]
for name in test_names:
    print(repr(name).ljust(16), "->", validate_username(name))
````

### 8.7 Alignment and Padding Methods

````python
word = "Python"
width = 20
print("|" + word.center(width) + "|")
print("|" + word.ljust(width) + "|")
print("|" + word.rjust(width) + "|")
print("|" + word.center(width, "-") + "|")
print("42".zfill(6))   # "000042"
````

````python
# Real-world: formatted product table
print("Product".ljust(20) + "Price".rjust(10) + "Stock".rjust(10))
print("-" * 40)
products = [
    ("Python Book", 39.99, 150),
    ("Keyboard", 89.50, 42),
    ("Monitor", 349.00, 8),
    ("USB Cable", 9.99, 500),
]
for name, price, stock in products:
    print(name.ljust(20) + str(price).rjust(10) + str(stock).rjust(10))
````

## 9. f-Strings (Formatted String Literals)

### What Are f-Strings?

An **f-string** is a string prefixed with `f` that allows you to embed Python expressions **directly inside the string** using curly braces `{}`.

Introduced in Python 3.6, f-strings are the **standard, recommended way** to format strings.

### Syntax

```python
f"text {expression} more text"
```

| Part | Meaning |
|---|---|
| `f` prefix | Tells Python this is a formatted string |
| `{expression}` | Any valid Python expression -- variable, calculation, method call |
| Text outside `{}` | Treated as literal characters |

### Why f-Strings?

Before f-strings, Python used `%` formatting and `.format()`. f-strings are **shorter, more readable, and faster** than both.

````python
# Three ways to format -- f-string is the modern approach
name = "Alice"
age = 28

# Concatenation (verbose, requires str())
print("Hello, " + name + ". You are " + str(age) + " years old.")

# .format() (older approach)
print("Hello, {}. You are {} years old.".format(name, age))

# f-string (recommended)
print(f"Hello, {name}. You are {age} years old.")
````

````python
# Expressions inside f-strings -- not just variables
x = 10
y = 3
print(f"Sum:      {x + y}")
print(f"Product:  {x * y}")
print(f"Power:    {x ** y}")
print(f"Division: {x / y:.2f}")  # :.2f = 2 decimal places

# Method calls inside f-strings
name = "alice smith"
print(f"Formatted: {name.title()}")
````

````python
# Format specifiers -- controlling how values appear
price = 3.14159
big_number = 1234567
score = 0.8765
print(f"2 decimal places:   {price:.2f}")
print(f"4 decimal places:   {price:.4f}")
print(f"Thousands separator:{big_number:,}")
print(f"Percentage:         {score:.1%}")
print(f"Scientific:         {price:e}")

# Width and alignment
for word in ["apple", "banana", "cherry"]:
    print(f"|{word:<10}|{len(word):>5}|")
````

````python
# Real-world: formatted student report
student_name  = "Alice Johnson"
math_score    = 94
english_score = 87
science_score = 91
average = (math_score + english_score + science_score) / 3

sep = "=" * 38
print(sep)
print(f"Student Report: {student_name}")
print(sep)
print(f"  Mathematics:  {math_score}/100")
print(f"  English:      {english_score}/100")
print(f"  Science:      {science_score}/100")
print(f"  {chr(8212) * 20}")
print(f"  Average:      {average:.1f}/100")
print(sep)
````

````python
# Multiline f-string
name = "Bob"
order_id = "ORD-2024-001"
total = 149.99
receipt = f"""Order Confirmation
{"─" * 25}
Customer:  {name}
Order ID:  {order_id}
Total:     ${total:.2f}
{"─" * 25}
Thank you, {name}!
"""
print(receipt)
````

````python
# f-string debugging with = specifier (Python 3.8+)
x = 42
y = 7
print(f"{x = }")
print(f"{y = }")
print(f"{x + y = }")
print(f"{x * y = }")
````

## 10. Edge Cases and Common Beginner Mistakes

These are the mistakes beginners make most often when working with strings.

````python
# MISTAKE 1 -- IndexError from off-by-one
word = "Hello"
print("Valid indexes: 0 to", len(word)-1, "and", -len(word), "to -1")
# word[5] raises IndexError -- use word[-1] for last character
print("Last char (safe):", word[-1])
````

````python
# MISTAKE 2 -- Forgetting stop is exclusive in slicing
word = "Python"
print("word[0:3]:", word[0:3])  # positions 0, 1, 2 -- NOT 3
print("word[0:4]:", word[0:4])  # positions 0, 1, 2, 3
print()
print("To get first N characters: string[:N]")
print("First 3:", word[:3])  # Pyt
print("First 4:", word[:4])  # Pyth
````

````python
# MISTAKE 3 -- Not capturing the return value of a method
name = "alice"
name.upper()              # Result is discarded!
print("After name.upper() without capture:", name)  # Still "alice"

# Correct: capture the return value
upper_name = name.upper()
print("Correctly captured:", upper_name)

# Or reassign:
name = name.upper()
print("Reassigned:", name)
````

````python
# MISTAKE 4 -- Using find() result without checking for -1
text = "Hello, World!"

# DANGER: if find returns -1, using it as index gives the LAST character
position = text.find("xyz")
print("find result:", position)  # -1
# print(text[position])  # This would give "!" -- wrong!

# Correct: always check
if position == -1:
    print("Substring not found")
else:
    print("Found at position:", position)
````

````python
# MISTAKE 5 -- Building a string with += in a large loop
words = ["Python", "is", "great", "for", "all"]

# Creates a new string object every iteration (slow for large data):
result = ""
for word in words:
    result += word + " "
print("Works:", result.strip())

# Better: collect in a list, then join once:
result = " ".join(words)
print("Better:", result)
````

````python
# MISTAKE 6 -- Mixing string and int without conversion
age_string = "25"
# print(age_string + 1)  # TypeError: can only concatenate str to str

# Correct: convert first
age_number = int(age_string)
print("Age next year:", age_number + 1)

# f-strings handle numbers automatically -- no str() needed
score = 95
print(f"Your score: {score}")
````

## 11. How Strings Work Internally

### Unicode: The Foundation

Python stores each string as a sequence of **Unicode code points**. Every character -- from Latin letters to Chinese characters -- has a unique Unicode number. You can see this with `ord()` (character to number) and `chr()` (number to character).

````python
# Unicode code points
print("ord(A):", ord("A"))    # 65
print("ord(a):", ord("a"))    # 97
print("ord(0):", ord("0"))    # 48
print("ord(!):", ord("!"))    # 33
print()
print("chr(65):", chr(65))    # A
print("chr(97):", chr(97))    # a
print("chr(960):", chr(960))  # Greek letter pi
print()
# Uppercase letters have smaller code points than lowercase
print("A < a:", "A" < "a")    # True -- 65 < 97
````

````python
# String interning: Python may reuse identical string objects
a = "hello"
b = "hello"
print("a == b:", a == b)   # True -- same value
print("a is b:", a is b)   # Often True -- same object in memory
print()
print("Rule: always use == to compare string VALUES.")
print("Use is only for identity checks, which is rare.")
````

````python
# Slicing always creates a new object
original = "Hello, World!"
sliced = original[0:5]
print("original:", original)
print("sliced:  ", sliced)
print("Same object?", original is sliced)  # Always False
print("id(original):", id(original))
print("id(sliced):  ", id(sliced))
````

## 12. Best Practices

### String Formatting

- **Use f-strings** for almost all formatting -- readable and fast
- Avoid concatenating strings with `+` when building formatted output
- Use `.format()` only for dynamic format strings or Python < 3.6 compatibility

### String Building

- **Use `' '.join(list)`** to build strings from many parts
- `+=` in a loop is fine for a few concatenations, slow for thousands

### Input Handling

- Always `.strip()` user input before processing or comparing
- Use `.lower()` or `.upper()` before comparing user-entered strings
- Use `startswith()` and `endswith()` instead of slicing for prefix/suffix checks

### General

- Use `word[-1]` instead of `word[len(word)-1]`
- Use `in` for membership checks instead of `.find()` when you only need True/False
- Use raw strings `r'...'` for file paths and regular expressions

````python
# f-string vs concatenation
name = "Alice"
score = 94.5

# Avoid:
print("Name: " + name + " | Score: " + str(score))

# Prefer:
print(f"Name: {name} | Score: {score:.1f}")
````

````python
# "in" for membership, not find()
email = "alice@example.com"

# Verbose:
if email.find("@") != -1:
    print("Contains @ (verbose)")

# Clean:
if "@" in email:
    print("Contains @ (clean)")
````

````python
# join() for building strings from lists
words = ["the", "quick", "brown", "fox"]

# Avoid in large loops:
sentence = ""
for w in words:
    sentence += w + " "
print("Avoid:", sentence.strip())

# Prefer:
sentence = " ".join(words)
print("Prefer:", sentence)
````

Complete, realistic programs demonstrating string techniques in action.

````python
# Application 1 -- Email address validator
def validate_email(email):
    email = email.strip().lower()
    if len(email) == 0:
        return False, "Email cannot be empty"
    if "@" not in email:
        return False, "Missing @ symbol"
    parts = email.split("@")
    if len(parts) != 2:
        return False, "Invalid @ placement"
    local, domain = parts
    if len(local) == 0:
        return False, "Missing username before @"
    if "." not in domain:
        return False, "Domain must contain a dot"
    if domain.startswith(".") or domain.endswith("."):
        return False, "Domain cannot start or end with dot"
    return True, "Valid email"

test_emails = [
    "alice@example.com",
    "bob.smith@company.org",
    "invalid-email",
    "@nodomain.com",
    "two@@signs.com",
    "  alice@test.com  ",
]
for email in test_emails:
    valid, message = validate_email(email)
    status = "VALID  " if valid else "INVALID"
    print(f"{repr(email):<30} -> {status}: {message}")
````

````python
# Application 2 -- Text statistics analyzer
def analyze_text(text):
    if len(text.strip()) == 0:
        return "Empty text provided"
    words = text.split()
    sentences = text.replace("!", ".").replace("?", ".").split(".")
    sentences = [s.strip() for s in sentences if s.strip()]
    char_count = len(text)
    char_no_space = len(text.replace(" ", ""))
    word_count = len(words)
    sentence_count = len(sentences)
    vowels = "aeiouAEIOU"
    vowel_count = sum(1 for ch in text if ch in vowels)
    longest = max(words, key=len) if words else ""
    avg_word_len = char_no_space / word_count if word_count > 0 else 0
    sep = "=" * 35
    return f"""{sep}
Text Analysis Report
{sep}
Characters (total):     {char_count:>6}
Characters (no spaces): {char_no_space:>6}
Words:                  {word_count:>6}
Sentences:              {sentence_count:>6}
Vowels:                 {vowel_count:>6}
Longest word:           {longest:>6}
Avg word length:        {avg_word_len:>6.1f}
{sep}"""

sample = "Python is powerful. It is easy to learn! Many developers love Python."
print(analyze_text(sample))
````

````python
# Application 3 -- Name formatter and initials generator
def process_name(raw_name):
    name = raw_name.strip()
    if not name:
        return {"error": "Empty name"}
    name = " ".join(name.split())  # normalize whitespace
    parts = name.split()
    formatted = name.title()
    initials = "".join(part[0].upper() + "." for part in parts)
    first = parts[0].title()
    last = parts[-1].title() if len(parts) > 1 else ""
    return {"full": formatted, "first": first, "last": last, "initials": initials}

test_names = ["  alice   smith  ", "JOHN MICHAEL JONES", "carol", "   ", "robert FITZGERALD"]
for raw in test_names:
    result = process_name(raw)
    if "error" in result:
        print("Input", repr(raw).ljust(26), "-> Error:", result["error"])
    else:
        print("Input", repr(raw).ljust(26), "-> Full:", result["full"].ljust(22), "Initials:", result["initials"])
````

## 14. Summary

### Strings as Sequences

- An ordered sequence of characters, each stored at a numbered index
- Indexes start at 0 from the left, -1 from the right

### Indexing and Slicing

- `string[i]` -- single character at position `i`
- `string[start:stop:step]` -- substring; `stop` is exclusive
- Slicing never raises IndexError; indexing does
- `string[::-1]` reverses any string

### Escape Characters

- `\n` (newline), `\t` (tab), `\\` (backslash), `\"` / `\'` (quotes)
- Raw strings `r'...'` treat all backslashes literally

### Multiline Strings

- Triple quotes `"""..."""` create strings spanning multiple lines
- Actual newlines inside are preserved in the string

### Immutability

- Strings cannot be changed in place -- all operations return new strings
- Reassigning a variable is NOT the same as modifying the string

### Key String Methods

- `upper()`, `lower()`, `title()` -- case conversion
- `strip()`, `lstrip()`, `rstrip()` -- whitespace removal
- `find()`, `count()`, `startswith()`, `endswith()` -- searching
- `split()`, `join()` -- the most important pair
- `replace()` -- substitution
- `isdigit()`, `isalpha()`, `isalnum()` -- content checking

### f-Strings

- `f'...'` embeds any expression in `{}`
- `:.2f` (decimals), `:,` (thousands), `:.1%` (percent), `:<10` (width)
- `{x=}` for debugging (Python 3.8+)
