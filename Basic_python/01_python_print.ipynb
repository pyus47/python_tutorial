<h2 style="color:#e8a838;font-family:'Segoe UI',sans-serif;#e8a838;padding-left:12px;">
What is Python?
</h2>

**Python** is a general-purpose, high-level programming language created by **Guido van Rossum** and first released in **1991**. The name is a nod to the British comedy group *Monty Python* — a hint from its creator that programming should be enjoyable, not frustrating.

The word **"high-level"** is important. It means Python handles the messy low-level details of your computer — memory management, hardware instructions, byte allocation — so that *you* can focus entirely on solving problems.

Here's why Python reads almost like English. Compare printing text in different languages:

| Language | Code to print "Hello, world!" |
|----------|-------------------------------|
| **Python** | `print("Hello, world!")` |
| Java | `System.out.println("Hello, world!");` |
| C++ | `std::cout << "Hello, world!" << std::endl;` |

Python removes curly braces, semicolons, and type declarations. The result: code that closely mirrors how you'd describe a problem out loud.

---

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">Four core properties of Python</h3>

| Property | What it means | Practical impact |
|----------|---------------|-----------------|
| **Interpreted** | Runs line-by-line, no compile step | Write → run → see results instantly |
| **Dynamically typed** | No need to declare variable types | `x = 5` just works |
| **Garbage collected** | Memory managed automatically | You never manually free memory |
| **Batteries included** | Huge standard library built-in | `import json`, `import os` — it's already there |

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">Understanding the .py file</h3>

When you run `python hello.py`, here is exactly what happens:

1. The OS finds the Python executable (because of PATH)
2. Python opens and reads `hello.py` as plain text
3. The **lexer** breaks the text into tokens (keywords, strings, operators)
4. The **parser** builds an Abstract Syntax Tree (AST) from the tokens
5. The **compiler** turns the AST into **bytecode** (`.pyc` files in `__pycache__/`)
6. The **Python Virtual Machine (PVM)** executes the bytecode line by line

You never see steps 3–6 happen — they're all automatic. This is what "interpreted" means in practice.

<h2 style="color:#e8a838;font-family:'Segoe UI',sans-serif; solid #e8a838;padding-left:12px;">
The Python REPL
</h2>

**REPL** stands for **Read-Eval-Print Loop**. It's an interactive Python session where you type one piece of code, Python executes it immediately, prints the result, then waits for the next input. Think of it as a direct conversation with Python.

---

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">Why the REPL matters — especially for learning</h3>

The REPL is your fastest learning tool. Whenever you're curious about something:

- *"What does this function return?"*
- *"What type is this variable?"*
- *"Does this math work the way I think it does?"*

You don't need to create a file, save it, and run it. Open the REPL, type your question in code, get an answer in under 2 seconds. Senior Python developers use the REPL daily.

---

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">REPL vs .py script — key difference</h3>

| Behaviour | REPL | .py script |
|-----------|------|-----------|
| Expressions auto-displayed |  Yes |  No — silently discarded |
| Statements persist across lines |  Yes |  Yes |
| Best for | Quick experiments, exploration | Saving and reusing code |
| Start with | `python3` in terminal | `python3 hello.py` |
| Exit with | `exit()` or `Ctrl+D` | Script ends automatically |

**The crucial difference**: In the REPL, typing `2 + 2` shows `4` automatically. In a `.py` file, `2 + 2` on its own line does absolutely nothing — you must wrap it in `print(2 + 2)` to see any output.

---

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">Opening the REPL</h3>

In your terminal:

```bash
# macOS / Linux
python3

# Windows
python
```

You'll see the `>>>` prompt — Python is ready for input. To exit: type `exit()` and press Enter, or press `Ctrl+D`.

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">REPL behaviour — run these cells one at a time</h3>

````python
# In a .py file (like this notebook cell), expressions are NOT auto-displayed
# You MUST use print() to see output

2 + 2          # ← silent in a script! Nothing shows up
"hello"        # ← silent in a script!

# Only print() produces visible output in a script
print(2 + 2)           # → 4
print("hello")         # → hello
print(type(42))        # → <class 'int'>
print(type("hello"))   # → <class 'str'>
````

````python
# The REPL's auto-display behaviour is simulated here for teaching
# In a real REPL session, each line below would auto-display its result

expressions = [
    2 + 2,
    10 * 3.5,
    "hello".upper(),
    len("Python"),
    [1, 2, 3],
    type(True),
]

print("What the REPL would show for each expression:")
print("-" * 40)
for expr in expressions:
    print(repr(expr))
````

<h2 style="color:#e8a838;font-family:'Segoe UI',sans-serif;  #e8a838;padding-left:12px;">
<code>print()</code> · Your First and Most Powerful Tool
</h2>

`print()` is a **built-in function** that sends output to the *standard output* — which in most cases means your terminal screen. It is the first function every Python programmer learns, and the one they use every single day, even after years of experience.

But `print()` is far more capable than most beginners realize. Let's learn it completely.

---

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">The complete signature</h3>

```python
print(*objects, sep=' ', end='\n', file=sys.stdout, flush=False)
```

Every argument except `*objects` has a **default value** — meaning `print("hello")` works perfectly without specifying anything else. But each parameter gives you fine-grained control when you need it.

---

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">Parameter 1 — <code>*objects</code> (the things to print)</h3>

The `*` means it accepts **zero or more arguments** of any type. Python internally calls `str()` on every object before printing, converting integers, floats, lists — anything — to text automatically.

```python
print()                      # no arguments → blank line
print("hello")               # one string
print("x =", 42)             # string + integer (auto-converted)
print(1, 2, 3, "go")         # multiple mixed values
print([1, 2, 3])             # a list
print(True, None, 3.14)      # bool, None, float
```

---

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">Parameter 2 — <code>sep</code> (the separator)</h3>

`sep` is the string inserted **between** each object when you pass multiple values. Default is a single space `' '`.

```python
print("a", "b", "c")           # → a b c        (default space)
print("a", "b", "c", sep="-")  # → a-b-c
print("a", "b", "c", sep="")   # → abc
print("a", "b", "c", sep=", ") # → a, b, c
print("a", "b", sep="\n")      # → a            (each on own line)
                                #   b
```

**Why this matters**: `sep` lets you format output precisely without string concatenation. Building a date like `2024-03-15`? Pass the parts as separate arguments with `sep="-"` — cleaner than `"2024" + "-" + "03" + "-" + "15"`.

---

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">Parameter 3 — <code>end</code> (what comes after)</h3>

`end` is what gets printed at the very **end**, after all objects. Default is `'\n'` (a newline character) — which is why each `print()` call moves to a new line.

Change it to `""` to stay on the same line. Change it to `" "` to separate print calls with a space.

```python
print("Hello", end=" ")
print("World")          # → Hello World  (same line!)

print("Loading", end="...")   # → Loading...
```

---

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">Parameter 4 — <code>file</code> (where output goes)</h3>

`file` controls **where** the output is written. Default is `sys.stdout` (your terminal). You can redirect it to a real file, or to `sys.stderr` for error output.

```python
import sys

# Write to error stream (appears red in many terminals)
print("Something went wrong!", file=sys.stderr)

# Write directly to a file
with open("output.txt", "w") as f:
    print("This goes into the file", file=f)
    print("So does this", file=f)
```

---

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">Parameter 5 — <code>flush</code> (force immediate output)</h3>

Python **buffers** output for performance — it collects output and sends it in chunks rather than one character at a time. When `flush=True`, Python writes immediately without waiting for the buffer to fill.

**When you need it**: Progress indicators, real-time logging, any situation where you need to see partial output the moment it's generated.

```python
import time

for i in range(5):
    print(f"Step {i+1}/5", end="\r", flush=True)   # overwrites same line
    time.sleep(0.5)
print("Complete!     ")   # final message
```

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;"> Code examples — basic print()</h3>

````python
# ── Basic print() usage ──────────────────────────────────────────────────────

# 1. Single value — the most common use
print("Hello, Python!")

# 2. Multiple values — Python puts a space between each by default
name = "Fatima"
age  = 22
print("Name:", name, "| Age:", age)

# 3. Printing different types — print() converts all to str automatically
print("Score:", 98.5, "out of", 100)

# 4. Empty print() — prints a blank line (great for spacing output)
print()
print("After the blank line above")

# 5. Printing Python objects — lists, dicts, booleans, None
print([1, 2, 3])
print({"key": "value"})
print(True, False, None)
````

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">Code examples  sep parameter</h3>

````python
# ── sep — controlling the separator ─────────────────────────────────────────

# Default sep is a single space
print("2024", "03", "15")               # → 2024 03 15

# Build a date format
print("2024", "03", "15", sep="-")      # → 2024-03-15

# Build a file path
print("home", "user", "documents", "project.py", sep="/")

# No separator — items run together
print("Hello", "World", sep="")         # → HelloWorld

# Comma + space separator — like a list
fruits = ["Apple", "Banana", "Cherry"]
print(*fruits, sep=", ")                # unpack list with * and use sep

# Newline separator — each item on its own line
print("Step 1", "Step 2", "Step 3", "Step 4", sep="\n")
````

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;"> Code examples  end parameter</h3>

````python
# ── end — controlling line endings ──────────────────────────────────────────

# Default: each print ends with \n
print("Line 1")
print("Line 2")       # separate lines

print()   # blank line for spacing

# end="" — stay on the same line
print("Processing", end="")
print("...", end="")
print(" Done!")        # all three on one line: Processing... Done!

print()

# end=" " — print with trailing space instead of newline
for word in ["Python", "is", "awesome"]:
    print(word, end=" ")
print()    # move to next line at the end

# Useful for counting or progress — overwrite the same line
````

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">Code examples  f-strings with print()</h3>

````python
# ── f-strings — the modern way to embed values in strings ───────────────────
# f-strings (formatted string literals) were introduced in Python 3.6
# Prefix the string with f and embed any expression inside {}

name    = "Aisha"
score   = 94.7
subject = "Mathematics"
rank    = 3

# Basic variable embedding
print(f"Student: {name}")
print(f"Subject: {subject} | Score: {score}%")

# Expressions inside {} — not just variables
print(f"Score doubled: {score * 2:.1f}%")    # .1f = 1 decimal place
print(f"Is passing (>= 50)? {score >= 50}")

# Method calls inside {}
print(f"Name in uppercase: {name.upper()}")
print(f"Name length: {len(name)} characters")

# Alignment and padding — useful for tables
````

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;"> Code examples file and flush parameters</h3>

<h2 style="color:#e8a838;font-family:'Segoe UI',sans-serif; solid #e8a838;padding-left:12px;">
Edge Cases & Surprising Behaviours
</h2>

These are the situations that trip up beginners. Understanding them now will save you a lot of confusing debugging sessions later.

---

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">Edge case 1 — print() returns None, not the string</h3>

`print()` is a function. In Python, all functions return something. `print()` always returns `None` — its purpose is the *side effect* (displaying text), not a return value.

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">Edge case 2 — Printing None prints the word "None"</h3>

If a variable holds `None`, printing it shows the literal text `None`. This commonly happens when you try to print the return value of a function that doesn't have a `return` statement.

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">Edge case 3 — Script vs REPL auto-display</h3>

In the REPL, typing `2 + 2` shows `4`. In a `.py` script, that same line produces zero output. Always use `print()` in scripts.

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">Edge case 4 — Escape sequences in strings</h3>

Certain characters inside strings have special meaning when preceded by a backslash `\`:

| Sequence | Meaning | Example output |
|----------|---------|----------------|
| `\n` | Newline | moves to next line |
| `\t` | Tab | horizontal indent |
| `\\` | Literal backslash | `\` |
| `\"` | Literal double quote | `"` |
| `\'` | Literal single quote | `'` |
| `\r` | Carriage return | overwrites current line |

Use a **raw string** (`r"..."`) to treat backslashes literally — crucial for Windows file paths.

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;"> Edge cases in code</h3>

````python
# ── Edge case 1: print() returns None ──────────────────────────────────────

result = print("This prints, but returns...") 
print(f"Return value of print() is: {repr(result)}")    # → None 

print()

# ── Edge case 2: Printing None ───────────────────────────────────────────────

nothing = None
print(nothing)         # → None   (the text "None", not an error)

def greet(name):
    print(f"Hello, {name}")
    # No return statement → implicitly returns None

return_val = greet("Ali")
print(f"greet() returned: {repr(return_val)}")   # → None

print()

# ── Edge case 3: Escape sequences ────────────────────────────────────────────

print("Tab:	here")                   # 	 = tab
print("NewLine")                    # = newline inside string
print("Quote: hello")            # " = literal quote inside double-quoted string
print(r"Windows path: C:/Users/name/docs")   # r prefix = raw string, no escaping

print()

# ── Edge case 4: Unicode works natively in Python 3 ──────────────────────────

print("Arabic: مرحبا بالعالم")
print("Urdu: پائتھن سیکھیں")
print("Math: π = 3.14159  |  √2 ≈ 1.41421")
print("Symbols: ✓  ✗  ★  →  ←")
print("\u2090")
````

````python
# ── Edge case 5: print() with sep="" vs string concatenation ─────────────────

# These produce identical output but the print approach is often cleaner
first = "Hello"
last  = "World"

# String concatenation — need to convert non-strings manually
combined = first + " " + last
print(combined)

# Using print with sep — no manual conversion needed
print(first, last)         # same output

# Practical: printing numbers without spaces
print(1, 0, 0, 1, 0, 1, sep="")    # → 100101  (like binary digits)

print()

# ── Edge case 6: Very long print() and line continuation ─────────────────────

# You can split long print statements across lines with print("This is a very long message that "
print( "spans multiple lines in the source "
      "but prints as a single line.")    # Python joins adjacent string literals
````

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">Exercise 3 — Solution</h3>

````python
# Exercise 3: Same-line printing with end=

# Approach 1: loop with end=
for i in range(1, 6):
    if i < 5:
        print(i, end=" → ")
    else:
        print(i)    # last number — use default newline

# Approach 2: using sep with unpacking (more elegant)
print(*range(1, 6), sep=" → ")

# Approach 3: building the string manually
parts = [str(i) for i in range(1, 6)]
print(" → ".join(parts))
````

<h2 style="color:#e8a838;font-family:'Segoe UI',sans-serif; solid #e8a838;padding-left:12px;">
Section 8 — Real-World Use Case
</h2>

<h3 style="color:#5ec4a8;font-family:'Segoe UI',sans-serif;">CLI Deployment Status Reporter</h3>

**Scenario**: You've written a Python script that deploys services to a server. The team needs to see a clean, readable terminal output showing each deployment step as it happens — service name, version, status, and timing. This is a simplified version of the kind of CLI tooling used at companies like Netflix, Shopify, and every software team that does automated deployments.

**Concepts used from this topic:**
- `print()` with `end=""` to build a line progressively
- `flush=True` to display partial lines immediately  
- f-strings for dynamic, aligned output
- `sep` for formatted separators
- `"=" * N` string multiplication for borders
- `file=sys.stderr` for error messages

````python
# ── Real-World Use Case: CLI Deployment Status Reporter ─────────────────────
import time
import sys

# Services to deploy: (name, version, expected_duration_seconds)
services = [
    ("api-gateway",      "v2.3.1", 0.6),
    ("auth-service",     "v1.8.0", 0.4),
    ("user-service",     "v3.1.2", 0.5),
    ("database-proxy",   "v4.0.2", 0.8),
    ("notification-svc", "v1.2.0", 0.3),
]

WIDTH = 52

# ── Header ───────────────────────────────────────────────────────────────────
print("=" * WIDTH)
print(f"{'  DEPLOYMENT LOG':^{WIDTH}}")
print(f"{'  Python Mastery Series — Topic 01':^{WIDTH}}")
print("=" * WIDTH)
print()

deployed = 0
failed   = 0

# ── Deploy each service ───────────────────────────────────────────────────────
for name, version, duration in services:

    # Print service name first — end="" keeps cursor on same line
    print(f"  Deploying {name:<22}", end="", flush=True)

    # Simulate deployment work
    time.sleep(duration)

    # All deployments succeed in this demo
    success = True

    if success:
        print(f"[ {version} ]   OK")
        deployed += 1
    else:
        print(f"[ {version} ]   FAILED", file=sys.stderr)
        failed += 1

# ── Summary ───────────────────────────────────────────────────────────────────
print()
print("=" * WIDTH)
print(f"  Deployed : {deployed} services")
if failed > 0:
    print(f"  Failed   : {failed} services", file=sys.stderr)
print(f"  Total    : {len(services)} services processed")
print("=" * WIDTH)
print()
print(f"  All {deployed} services deployed successfully.")
print("=" * WIDTH)
````
