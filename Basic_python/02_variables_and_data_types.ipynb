## 1. Intuition and Motivation

### The Problem This Concept Solves

Imagine you are writing a program that calculates the total bill at a restaurant. The bill might be `47.85`. The customer's name might be `"Alice"`. The number of guests might be `4`. And you might want to track whether a discount was applied: `True` or `False`.

Each of these pieces of information is fundamentally different in nature:

- A name is made of letters.
- A price has a decimal point.
- A count is a whole number.
- A yes/no flag is either true or false.

A computer needs to know **what kind of data** it is dealing with because the operations it can perform depend on the kind. You can multiply a price by 2 to double it, but multiplying a customer's name by 2 does not make sense mathematically.

### Why Data Types Exist

Data types exist to:

1. **Tell Python how to store the data in memory** — a whole number takes different space than a decimal or a sentence.
2. **Tell Python what operations are valid** — you can divide two numbers, but you cannot divide two names.
3. **Prevent errors** — if Python knows a value is text, it will warn you before you try to do arithmetic on it.

### Real-World Analogy

Think of data types like **different kinds of containers in a kitchen**:

- A **measuring cup** (float) holds liquid with precise amounts like `2.5` cups.
- An **egg carton** (int) holds whole items — you cannot have `2.7` eggs.
- A **label** (str) holds text — the name written on a jar.
- A **light switch** (bool) is either ON or OFF, nothing in between.

Just as you would not pour soup into an egg carton, Python needs to know which container holds which kind of value so it can handle it correctly.

## 2. What Is a Variable?

### The Core Idea

A **variable** is a named label that points to a value stored in your computer's memory. Think of it as a sticky note: the note has a name written on it, and it points to a piece of information.

When you write:

```python
age = 25
```

You are telling Python: "Create a sticky note called `age` and stick it onto the value `25`."

Later in your program, whenever you write `age`, Python goes and fetches the value `25` from memory.

### Why Variables Matter

Without variables, you would have to repeat every value every time you use it. If a product's price changes, you would have to hunt through every line of code and update it manually. With a variable, you change it in one place and it updates everywhere automatically.

### Basic Syntax

```
variable_name = value
```

- `variable_name` — The label you choose. You decide this.
- `=` — This is the **assignment operator**. It does NOT mean "equals" like in math. It means "store the value on the right into the name on the left".
- `value` — The actual data you want to remember.

There is no keyword like `var` or `let` needed in Python. You just write the name, the `=`, and the value.

````python
# Assigning values to variables
# Python creates the variable and stores the value in memory

city = "Lahore"
temperature = 38.5
population = 14000000
is_capital = False

# Retrieving the values by using the variable names
print(city)
print(temperature)
print(population)
print(is_capital)
````

````python
# Variables can be reassigned — the label simply moves to a new value
# This is one of Python's most powerful features

score = 10
print(score)   # Output: 10

score = 85     # The variable 'score' now points to 85 instead of 10
print(score)   # Output: 85

# The old value 10 is gone (Python will clean it up from memory automatically)
````

````python
# You can assign multiple variables on separate lines
product_name = "Laptop"
product_price = 75000
in_stock = True

print(product_name)
print(product_price)
print(in_stock)
````

````python
# Python also allows assigning multiple variables in a single line
# This is called multiple assignment

x, y, z = 10, 20, 30
print(x)
print(y)
print(z)
````

````python
# You can also assign the same value to multiple variables at once
# Useful when initializing several counters or flags to the same starting value

a = b = c = 0
print(a)
print(b)
print(c)
````

## 3. The Four Core Data Types

Python has four fundamental built-in data types that every beginner must master. Everything else in Python is built on top of these.

| Type | Python Keyword | Example | Represents |
|------|---------------|---------|------------|
| Integer | `int` | `42`, `-7`, `0` | Whole numbers |
| Float | `float` | `3.14`, `-0.5`, `2.0` | Decimal numbers |
| String | `str` | `"hello"`, `'Python'` | Text / characters |
| Boolean | `bool` | `True`, `False` | Yes/No, On/Off |

---

### 3.1 Integer (`int`)

An integer is a **whole number** — positive, negative, or zero — with no decimal point.

**Where you use it in real life:**
- Number of students in a class
- Age of a person
- Quantity of items in a cart
- Page number of a book

**What Python can do with integers:**
- Add, subtract, multiply, divide
- Compare (`>`, `<`, `==`)
- Use as index positions (covered in later lessons)

````python
# Basic integer examples
age = 22
year = 2024
below_zero = -15
zero = 0

print(age)
print(year)
print(below_zero)
print(zero)
````

````python
# Real-world example: Tracking items in a shopping cart
items_in_cart = 5
items_removed = 2
remaining_items = items_in_cart - items_removed

print(remaining_items)   # Output: 3
````

````python
# Python integers have NO size limit — they grow as large as needed
# Unlike some languages (like C or Java), Python handles huge numbers natively

# A very large integer — Python handles it perfectly
national_debt = 99999999999999999999999
print(national_debt)
````

````python
# You can use underscores inside large numbers for readability
# Python ignores the underscores — they are only for humans reading the code

population = 1_400_000_000    # 1.4 billion (same as 1400000000)
disk_space = 256_000_000_000  # 256 GB in bytes

print(population)
print(disk_space)
````

---
### 3.2 Float (`float`)

A float (short for **floating-point number**) is a number that has a decimal point. The word "floating" refers to how the decimal point can "float" to different positions.

**Where you use it in real life:**
- Prices and money (`99.99`)
- Temperatures (`36.6`)
- GPS coordinates (`31.5497`)
- Percentages calculated as decimals (`0.85` for 85%)
- Scientific measurements

**Important distinction from int:**
Even `2.0` is a float, not an integer. The presence of a decimal point is what makes it a float.

````python
# Basic float examples
price = 499.99
pi = 3.14159
body_temperature = 98.6
negative_float = -7.5
whole_as_float = 2.0     # This is still a float, NOT an integer

print(price)
print(pi)
print(body_temperature)
print(negative_float)
print(whole_as_float)
````

````python
# Real-world example: Calculating a discounted price
original_price = 1200.00
discount_rate = 0.15       # 15% discount
discount_amount = original_price * discount_rate
final_price = original_price - discount_amount

print(discount_amount)    # 180.0
print(final_price)        # 1020.0
````

````python
# Edge case: Division in Python ALWAYS returns a float
# Even when dividing two integers that divide evenly

result = 10 / 2    # You might expect 5 (int), but Python gives 5.0 (float)
print(result)      # Output: 5.0

# This is important to remember — the / operator always produces a float
````

````python
# Advanced: Floating-point precision — a subtle but important reality
# Computers store floats in binary, which can cause tiny precision issues

result = 0.1 + 0.2
print(result)   # You might expect 0.3, but you get 0.30000000000000004

# This is NOT a Python bug — it is a fundamental property of how all
# computers represent decimal numbers in binary (IEEE 754 standard).
# For most everyday code, this does not matter.
# For financial apps, the 'decimal' module is used instead.
````

---
### 3.3 String (`str`)

A string is a **sequence of characters** — letters, digits, spaces, punctuation, or any symbol — enclosed in quotes.

The word "string" comes from the idea of characters **strung together** like beads on a necklace.

**Where you use it in real life:**
- Names, addresses, messages
- File paths and URLs
- Any text a program reads or displays
- User input is always received as a string

**Two ways to create a string:**
- Single quotes: `'hello'`
- Double quotes: `"hello"`

Both are identical in Python. The convention is to be consistent — pick one style and stick with it.

````python
# Basic string examples
first_name = "Ali"
last_name = 'Ahmed'
greeting = "Hello, World!"
empty_string = ""       # A string with no characters — still a valid string
space_string = " "      # A string containing only a space

print(first_name)
print(last_name)
print(greeting)
print(empty_string)     # Prints a blank line
print(space_string)
````

````python
# Strings can contain digits — but they are TEXT, not numbers
# The digit '5' as a string is completely different from the number 5

phone_number = "0300-1234567"    # This should be a string, not an int
zip_code = "54000"               # Zip codes are strings (may start with 0)
pin_code = "4782"

print(phone_number)
print(zip_code)
print(pin_code)
````

````python
# String concatenation: joining two strings together using +
first_name = "Fatima"
last_name = "Khan"
full_name = first_name + " " + last_name    # The " " adds a space between them

print(full_name)   # Output: Fatima Khan
````

````python
# Multi-line strings use triple quotes (three single or double quotes)
# Useful for long text, addresses, or blocks of descriptive content

address = """House 12,
Street 4, Block B,
Gulberg III,
Lahore"""

print(address)
````

````python
# When your string contains a quote character, use the opposite type of quotes
# OR use a backslash (escape character) to include it

sentence1 = "He said, 'Python is amazing!'"
sentence2 = 'She replied, "I agree!"'
sentence3 = "It\'s a beautiful day."    # Backslash escapes the apostrophe

print(sentence1)
print(sentence2)
print(sentence3)
````

---
### 3.4 Boolean (`bool`)

A boolean has exactly **two possible values**: `True` or `False`. Nothing else.

The word "boolean" comes from George Boole, the mathematician who developed Boolean algebra — the mathematical foundation of all computer logic.

**Why booleans exist:**
Every decision a computer makes ultimately comes down to True/False. Is the user logged in? Is the temperature above the limit? Has the file been saved? All of these are yes/no questions.

**Critical syntax rule:**
- `True` — capital T, no quotes
- `False` — capital F, no quotes
- `true` or `false` (lowercase) will cause an error in Python
- `"True"` (with quotes) is a string, NOT a boolean

````python
# Basic boolean examples
is_logged_in = True
has_premium = False
is_adult = True
is_weekend = False

print(is_logged_in)
print(has_premium)
print(is_adult)
print(is_weekend)
````

````python
# Booleans are produced by comparison operations
# Any time you compare two values, Python returns a boolean

print(10 > 5)    # True  — 10 is greater than 5
print(10 < 5)    # False — 10 is not less than 5
print(10 == 10)  # True  — 10 equals 10
print(10 == 11)  # False — 10 does not equal 11
print(10 != 5)   # True  — 10 is not equal to 5
````

````python
# Booleans have numeric values in Python
# True equals 1, False equals 0
# This is a deep Python design decision and becomes very useful later

print(True + True)    # 1 + 1 = 2
print(True + False)   # 1 + 0 = 1
print(False + False)  # 0 + 0 = 0

# This means bool is actually a subtype of int in Python
````

````python
# Real-world example: Checking user access level
user_age = 20
has_id = True

is_eligible = user_age >= 18    # Produces True or False based on comparison
can_enter = is_eligible         # We could also combine: is_eligible and has_id

print(is_eligible)   # True
print(can_enter)     # True
````

## 4. The `type()` Function

### What Is It?

`type()` is a built-in Python function that tells you the data type of any value or variable. It answers the question: "What kind of thing is this?"

### Syntax

```python
type(value)
```

You pass in a value (or a variable), and Python returns the type of that value.

### Why It Is Useful

When you are debugging code — trying to find why something is not working — `type()` is one of the most valuable tools. It instantly shows you what kind of data you are actually working with, which is often different from what you expected.

````python
# Using type() on direct values
print(type(42))        # <class 'int'>
print(type(3.14))      # <class 'float'>
print(type("hello"))   # <class 'str'>
print(type(True))      # <class 'bool'>
````

````python
# Using type() on variables
name = "Hamza"
score = 95
gpa = 3.87
graduated = True

print(type(name))       # <class 'str'>
print(type(score))      # <class 'int'>
print(type(gpa))        # <class 'float'>
print(type(graduated))  # <class 'bool'>
````

````python
# A tricky case: numbers in quotes are strings, not numbers
# type() helps you catch this mistake

a = 100      # This is an integer
b = "100"    # This is a string — it LOOKS like a number but it is text

print(type(a))   # <class 'int'>
print(type(b))   # <class 'str'>

# If you try to add them:
# a + b would cause a TypeError because Python cannot add int and str
````

````python
# Checking the type of the number 2.0 vs 2
# This is a classic beginner confusion

x = 2
y = 2.0

print(type(x))   # <class 'int'>
print(type(y))   # <class 'float'>

# They represent the same mathematical value but are different data types
print(x == y)    # True — same value
print(type(x) == type(y))  # False — different types
````

````python
# Professional debugging pattern: print both value and type together
# This is extremely useful when tracking down bugs

user_input = "42"   # Imagine this came from a form field
print("Value:", user_input, "| Type:", type(user_input))

# This tells you: the value is 42, but it is stored as a string
# You would need to convert it to int before doing math with it
````

## 5. Dynamic Typing

### What Is Dynamic Typing?

In Python, you **never need to declare a variable's type**. Python figures out the type automatically based on the value you assign. This is called **dynamic typing**.

Compare this to languages like Java or C++, where you must explicitly declare the type:

```java
// Java — you must declare the type
int age = 25;
String name = "Alice";
```

```python
# Python — type is inferred automatically
age = 25
name = "Alice"
```

### The Deeper Meaning of "Dynamic"

"Dynamic" means the type can **change at runtime** — a variable that holds an integer now can hold a string later. The variable itself has no fixed type; the type belongs to the **value**, not the variable.

Think of the variable as an empty box. The box can hold any item. What matters is what is currently inside the box, not what the box itself is labeled.

````python
# Python infers the type from the value — no declaration needed
x = 10          # Python sees an integer
print(type(x))  # <class 'int'>

x = 10.5        # Now x holds a float
print(type(x))  # <class 'float'>

x = "ten"       # Now x holds a string
print(type(x))  # <class 'str'>

x = True        # Now x holds a boolean
print(type(x))  # <class 'bool'>
````

````python
# The TYPE belongs to the VALUE, not the variable name
# Every value in Python carries its type information with it

a = 5       # The value 5 knows it is an int
b = a       # b now points to the same value 5

print(type(a))  # <class 'int'>
print(type(b))  # <class 'int'>

# Both variables point to the same integer value 5
````

````python
# Benefit of dynamic typing: quick, readable code
# No boilerplate needed to declare and type every variable

# In a data processing script, you might do this:
raw_data = "2024-03-15"    # String — the date as text from a file
count = 0                   # Integer — a counter starting at zero
average = 0.0               # Float — will hold a calculated average
is_valid = False            # Boolean — a status flag

# Python handles all of this without a single type declaration
print(type(raw_data))
print(type(count))
print(type(average))
print(type(is_valid))
````

````python
# The risk of dynamic typing: you might accidentally change a type
# This is a common bug in larger programs

price = 199     # You meant this as an integer (price in whole units)
print(type(price))   # <class 'int'>

# Many lines later, by accident:
price = "199"   # You accidentally reassigned it as a string
print(type(price))   # <class 'str'>

# Now any math you do with 'price' will fail
# This is why good variable naming and careful coding habits matter
````

## 6. Naming Conventions

### Why Naming Matters

A variable name is a communication tool. Code is read by humans far more often than it is written. Good names make code self-documenting — readable without needing comments.

Consider these two versions:

```python
# Version A
x = 1200
y = 0.15
z = x - (x * y)

# Version B
original_price = 1200
discount_rate = 0.15
final_price = original_price - (original_price * discount_rate)
```

Version B is instantly understandable. Version A requires you to guess.

---

### Python's Official Naming Rules (PEP 8)

Python has an official style guide called **PEP 8** that all professional Python developers follow. Here are the rules for variable names:

**Rules (must follow — otherwise you get an error):**

| Rule | Valid Example | Invalid Example |
|------|-------------|----------------|
| Only letters, digits, and underscores | `score_1` | `score-1` (hyphen not allowed) |
| Cannot start with a digit | `value_1` | `1_value` (starts with digit) |
| Cannot be a Python keyword | `my_class` | `class` (reserved keyword) |
| Case-sensitive | `Score` and `score` are different | |

**Style conventions (should follow — no error, but bad practice if ignored):**

| Style | Name | Example |
|-------|------|--------|
| `snake_case` | Standard for variables | `first_name`, `item_count` |
| `UPPER_CASE` | Constants (values that never change) | `MAX_SPEED`, `PI` |
| `CamelCase` | Class names (covered later) | `BankAccount`, `UserProfile` |

````python
# Valid variable names — follow snake_case convention
first_name = "Zara"
last_name = "Malik"
age = 28
account_balance = 50000.75
is_active = True

print(first_name)
print(last_name)
print(age)
print(account_balance)
print(is_active)
````

````python
# Constants: by convention, use ALL_CAPS to signal a value that should not change
# Python does not enforce this — it is a communication signal to other programmers

PI = 3.14159265
SPEED_OF_LIGHT = 299_792_458   # meters per second
MAX_RETRIES = 3
APP_VERSION = "2.4.1"

print(PI)
print(SPEED_OF_LIGHT)
print(MAX_RETRIES)
print(APP_VERSION)
````

````python
# Python is case-sensitive — these are FOUR different variables
name = "lowercase"
Name = "Capitalized"
NAME = "uppercase"
nAmE = "mixed"

print(name)
print(Name)
print(NAME)
print(nAmE)

# Best practice: never create variables that differ only in capitalization
# It makes code confusing and leads to hard-to-find bugs
````

````python
# Bad naming examples — valid Python but poor practice
# These are technically correct but communicate nothing useful

x = 95           # What does x represent?
temp = "Alice"   # Is this temperature, or temporary?
d = 7.5          # What is d?
n = 1000         # Is this a count, an ID, a threshold?

# Better naming:
exam_score = 95
customer_name = "Alice"
discount_percentage = 7.5
daily_visitors = 1000

print(exam_score)
print(customer_name)
````

````python
# Python reserved keywords — these CANNOT be used as variable names
# If you use them, Python will raise a SyntaxError

import keyword
print(keyword.kwlist)   # Shows all reserved keywords in Python
````

## 7. Type Conversion and Casting

### What Is Type Conversion?

Type conversion (also called **type casting**) is the process of converting a value from one data type to another.

**Why is this needed?**

Consider a very common scenario: a user types their age into a form field. That input always arrives as a **string** (text), even if they typed `25`. Before you can do any math with it — like checking if they are old enough — you must convert it to an integer.

### Two Types of Conversion

| Type | Description | Triggered by |
|------|-------------|-------------|
| **Implicit** | Python converts automatically without you asking | Python itself |
| **Explicit** | You tell Python to convert using a function | You, the programmer |

### Explicit Conversion Functions

| Function | Converts to | Example |
|----------|-------------|--------|
| `int()` | Integer | `int("42")` → `42` |
| `float()` | Float | `float("3.14")` → `3.14` |
| `str()` | String | `str(100)` → `"100"` |
| `bool()` | Boolean | `bool(0)` → `False` |

---

### 7.1 Implicit Conversion (Python Does It For You)

````python
# Implicit conversion: Python automatically promotes int to float
# when mixing int and float in arithmetic

a = 5      # int
b = 2.0    # float

result = a + b   # Python converts a (int) to 5.0 (float) automatically
print(result)        # 7.0
print(type(result))  # <class 'float'>

# Python chose float because float is more precise than int
# If the result were int, you would lose the decimal information
````

````python
# Another implicit conversion: True and False act as 1 and 0
# Python treats booleans as integers in arithmetic contexts

result = True + 5
print(result)        # 6   (True becomes 1)
print(type(result))  # <class 'int'>

result2 = False + 10.5
print(result2)       # 10.5  (False becomes 0, 0 + 10.5 = 10.5)
print(type(result2)) # <class 'float'>
````

### 7.2 Explicit Conversion — `int()`

````python
# int() converts other types to an integer

# String to int (only works if the string contains a valid whole number)
age_text = "25"
age_number = int(age_text)
print(age_number)        # 25
print(type(age_number))  # <class 'int'>
````

````python
# float to int — truncates the decimal part (does NOT round, just cuts off)
price = 99.99
whole_price = int(price)
print(whole_price)   # 99  — the .99 is REMOVED, not rounded

negative_float = -3.7
whole_negative = int(negative_float)
print(whole_negative)  # -3  — again, truncation toward zero, not floor
````

````python
# bool to int
print(int(True))   # 1
print(int(False))  # 0
````

````python
# What happens when int() fails
# If the string does not represent a valid integer, Python raises a ValueError

try:
    bad_conversion = int("hello")
except ValueError as e:
    print("Error:", e)

try:
    also_bad = int("3.14")    # int() cannot convert a decimal string directly
except ValueError as e:
    print("Error:", e)
````

### 7.3 Explicit Conversion — `float()`

````python
# float() converts other types to a float

# String to float
rate_text = "4.75"
rate = float(rate_text)
print(rate)         # 4.75
print(type(rate))   # <class 'float'>
````

````python
# int to float — just adds a .0
count = 5
count_as_float = float(count)
print(count_as_float)   # 5.0

# This matters when you want division to produce a precise decimal result
# (though in Python 3, / always returns float anyway)
````

````python
# A string containing a whole number CAN be converted to float
# (unlike int(), which cannot handle decimal strings)
x = float("100")     # Works — produces 100.0
y = float("3.14")    # Works — produces 3.14
print(x)
print(y)
````

### 7.4 Explicit Conversion — `str()`

````python
# str() converts any type to its string representation
# This is one of the most commonly used conversions

age = 28
gpa = 3.9
active = True

age_str = str(age)
gpa_str = str(gpa)
active_str = str(active)

print(age_str)       # "28"
print(gpa_str)       # "3.9"
print(active_str)    # "True"

print(type(age_str))     # <class 'str'>
print(type(gpa_str))     # <class 'str'>
````

````python
# Real-world example: Building a message that combines text and numbers
# You cannot concatenate (join) a string and a number directly

student_name = "Omar"
final_grade = 87

# This would cause a TypeError:
# message = "Student: " + student_name + " | Grade: " + final_grade

# Correct approach — convert int to str first
message = "Student: " + student_name + " | Grade: " + str(final_grade)
print(message)
````

### 7.5 Explicit Conversion — `bool()`

````python
# bool() converts other types to True or False
# Python has a clear rule:
# The following values are considered False (called "falsy"):
#   0, 0.0, "" (empty string), None, False
# Everything else is considered True (called "truthy")

print(bool(0))      # False — zero is falsy
print(bool(1))      # True  — any non-zero number is truthy
print(bool(-5))     # True  — negative numbers are also truthy
print(bool(0.0))    # False — zero float is falsy
print(bool(3.14))   # True  — non-zero float is truthy
````

````python
# bool() on strings
print(bool(""))        # False — empty string is falsy
print(bool("hello"))   # True  — any non-empty string is truthy
print(bool("0"))       # True  — the string "0" is NOT the number 0
print(bool("False"))   # True  — the string "False" is not the boolean False
````

## 8. Edge Cases and Common Mistakes

This section covers the most frequent mistakes beginners make. Understanding these will save you hours of frustration.

````python
# MISTAKE 1: Confusing = (assignment) with == (comparison)

x = 10      # CORRECT: assigns 10 to x

# x = 10 == 10  would set x to the RESULT of the comparison (True)
# That's not a bug, but it's confusing

# The difference:
x = 5       # Assignment — puts 5 into x
result = (x == 5)  # Comparison — checks if x equals 5, returns True

print(x)       # 5
print(result)  # True
````

````python
# MISTAKE 2: Trying to add a string and a number directly
# Python cannot add int and str — they are incompatible types

age = 20
message = "My age is "

# Wrong — raises TypeError
try:
    result = message + age
except TypeError as e:
    print("Error:", e)

# Correct — convert int to str first
result = message + str(age)
print(result)
````

````python
# MISTAKE 3: Converting a float string directly to int
# int() cannot handle decimal strings

price_text = "199.99"

# Wrong — raises ValueError
try:
    price = int(price_text)
except ValueError as e:
    print("Error:", e)

# Correct — convert to float first, then to int if needed
price = float(price_text)    # Step 1: string to float
price_whole = int(price)     # Step 2: float to int (truncates decimal)
print(price)
print(price_whole)
````

````python
# MISTAKE 4: Thinking "True" (string) and True (boolean) are the same

a = True          # Boolean
b = "True"        # String

print(type(a))    # <class 'bool'>
print(type(b))    # <class 'str'>
print(a == b)     # False — different types, even though they look the same

# bool("True") is True because any non-empty string is truthy
# but "True" is NOT the boolean True
print(bool(b))    # True — because the string is non-empty, NOT because it says "True"
````

````python
# MISTAKE 5: Using a Python keyword as a variable name

# These will all raise SyntaxError:
# class = "Mathematics"      # 'class' is a keyword
# for = 10                   # 'for' is a keyword  
# print = "hello"            # 'print' is a built-in function name (avoid redefining)

# Safe alternatives:
class_name = "Mathematics"
loop_count = 10
message_text = "hello"

print(class_name)
````

````python
# MISTAKE 6: Forgetting that division always returns a float
# A common source of unexpected float values

total = 10 / 2
print(total)        # 5.0  -- NOT 5
print(type(total))  # <class 'float'>

# If you need integer division (no decimal), use // (floor division)
total_int = 10 // 2
print(total_int)        # 5
print(type(total_int))  # <class 'int'>
````

````python
# MISTAKE 7: Shadowing built-in function names
# This does not raise an error immediately, but breaks your code silently

# Before shadowing:
print(type(42))    # Works normally

# After shadowing 'type' with an integer:
# type = 5           # Now 'type' refers to 5, not the built-in function
# print(type(42))    # This would raise TypeError: 'int' object is not callable

# NEVER name your variables: print, type, int, float, str, bool, len, input, etc.
# These are built-in names that you will need throughout your code

# Safe alternatives:
data_type = "integer"
user_input = "hello"
print(data_type)
````

## 9. How It Works Internally

### The Python Object Model

In Python, **everything is an object**. When you write `age = 25`, Python does three things behind the scenes:

1. **Creates an object in memory** — an integer object holding the value `25`, tagged with the type `int`.
2. **Assigns an ID to it** — every object has a unique memory address (accessible via `id()`).
3. **Binds the variable name** `age` to that object.

The variable is just a **name tag** (a reference/pointer). The actual data lives in memory as an object.

### Reference Semantics

When you write `b = a`, you are NOT creating a copy of the value. Both `a` and `b` now point to the **same object** in memory. For simple types (int, float, str, bool), this does not cause problems because these types are **immutable** — their value cannot be changed once created.

````python
# id() returns the memory address (unique identity) of an object
# This lets us see that variables are just labels pointing to objects

x = 100
y = x     # y points to the SAME object as x

print(id(x))          # Memory address of the object 100
print(id(y))          # Same address — both point to the same object
print(id(x) == id(y)) # True — they are the same object in memory
````

````python
# When you reassign, the variable points to a NEW object
# The original object is unchanged

x = 100
print(id(x))   # Address of object 100

x = 200        # x now points to a new object 200
print(id(x))   # Different address — a new object was created

# The object 100 still exists in memory momentarily
# Python's garbage collector will delete it when nothing references it anymore
````

````python
# Python caches small integers (-5 to 256) for performance
# This means variables holding these values share the same object in memory

a = 50
b = 50
print(id(a) == id(b))   # True — Python reuses the cached object for 50

# For large integers, Python creates new objects
c = 1000
d = 1000
print(id(c) == id(d))   # May be False — two separate objects for 1000
                         # (behavior may vary slightly between Python implementations)
````

````python
# Strings are also immutable — they cannot be changed in place
# When you "change" a string, Python creates a brand-new string object

name = "Ali"
print(id(name))    # Address of "Ali" object

name = "Ali Khan"  # A NEW string object is created; 'name' now points to it
print(id(name))    # Different address — completely new object

# The string "Ali" was not modified; it was replaced
````

## 10. Best Practices

### When to Use Each Type

| Use Case | Recommended Type | Reason |
|----------|-----------------|--------|
| Counting items | `int` | No decimals needed |
| Prices, measurements | `float` | Precision required |
| Names, messages, text | `str` | It is text |
| Yes/no flags, states | `bool` | Clearer than 0/1 |
| Phone numbers, IDs | `str` | No math needed; may have leading zeros |
| Percentages (0.85 = 85%) | `float` | More precise than storing as int |

### Key Guidelines

````python
# BEST PRACTICE 1: Choose the most appropriate type for the data

# Bad: storing a flag as an integer
is_admin = 1   # What does 1 mean? You have to remember the convention

# Good: using a boolean — self-documenting
is_admin = True   # Instantly clear to anyone reading the code
````

````python
# BEST PRACTICE 2: Store phone numbers, IDs, zip codes as strings

# Bad: phone as int
phone_bad = 3001234567   # Loses formatting information; no leading zero support

# Good: phone as string
phone_good = "0300-1234567"  # Preserves the format as-is

# Also: zip codes in some countries start with 0
zip_bad = 07001     # Python interprets this as octal (base-8) — bug!
zip_good = "07001"   # Perfectly preserved as text

print(zip_good)
````

````python
# BEST PRACTICE 3: Use descriptive, meaningful variable names

# Bad:
n = 150
p = 0.08
t = 12
r = n * p * t

# Good: the code explains itself
principal = 150
annual_interest_rate = 0.08
time_in_months = 12
total_interest = principal * annual_interest_rate * (time_in_months / 12)

print(total_interest)
````

````python
# BEST PRACTICE 4: Validate or check types when dealing with user input
# User input is always a string — verify before converting

raw_input = "42"  # Simulating user input

# Check if it can be safely converted before doing so
if raw_input.isdigit():
    number = int(raw_input)
    print("Converted successfully:", number)
else:
    print("That is not a valid whole number")
````

````python
# BEST PRACTICE 5: Keep variable types consistent throughout your code
# Avoid changing the type of a variable after assigning it

# Bad: type of 'value' changes unexpectedly
value = 100
# ... many lines later ...
value = "one hundred"   # Now 'value' is a string — confusing and bug-prone

# Good: use separate, clearly-named variables
count = 100
count_label = "one hundred"

print(count)
print(count_label)
````

## 12. Summary

<div style="background: #131415ff; border: 1px solid #040507ff; border-radius: 10px; padding: 28px 32px; margin: 10px 0;">

**Variables**
- A variable is a named label that points to a value in memory.
- Syntax: `name = value` — the `=` is assignment, not equality.
- Variables can be reassigned to new values at any time.

**The Four Core Data Types**
- `int` — whole numbers with no decimal point (`5`, `-3`, `0`).
- `float` — numbers with a decimal point (`3.14`, `99.99`, `2.0`).
- `str` — sequences of characters in quotes (`"hello"`, `'world'`).
- `bool` — either `True` or `False`, nothing else.

**The `type()` Function**
- Returns the data type of any value: `type(42)` returns `<class 'int'>`.
- Essential for debugging and verifying what type you are working with.

**Dynamic Typing**
- Python infers types automatically — no declaration needed.
- The type belongs to the *value*, not the variable name.
- A variable can be reassigned to a different type (but be careful).

**Naming Conventions**
- Use `snake_case` for variables: `first_name`, `account_balance`.
- Use `ALL_CAPS` for constants: `PI`, `MAX_SIZE`.
- Names must start with a letter or underscore, not a digit.
- Python is case-sensitive: `name`, `Name`, and `NAME` are three different variables.
- Never use Python keywords or built-in function names as variable names.

**Type Conversion and Casting**
- **Implicit**: Python promotes `int` to `float` automatically when needed.
- **Explicit**: Use `int()`, `float()`, `str()`, `bool()` to convert manually.
- `int("3.14")` fails — convert to `float` first, then to `int`.
- `str()` works on any type and is essential when building text messages.
- Values converted to `bool`: `0`, `0.0`, `""` are `False`; everything else is `True`.
