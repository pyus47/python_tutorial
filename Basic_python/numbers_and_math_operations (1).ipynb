## 2. Number Types in Python

Python has three primary numeric types. Understanding them is critical before doing any math.

| Type | Name | Example | Use Case |
|------|------|---------|----------|
| `int` | Integer | `5`, `-3`, `1000` | Counting, indexing, whole numbers |
| `float` | Floating-point | `3.14`, `-0.5`, `2.0` | Measurements, prices, decimals |
| `complex` | Complex number | `3+4j` | Advanced math, signal processing |

For 99% of beginner tasks, you will use `int` and `float`.

### How Python Stores These Internally

- **int**: Python integers have **unlimited precision**. Unlike many languages, Python `int` never overflows. `2 ** 1000` works perfectly.
- **float**: Stored as IEEE 754 double-precision binary (64-bit). Has a precision of about 15-17 significant digits. This is why `0.1 + 0.2` is not exactly `0.3` — we will explore this in depth later.
- **complex**: Written as `a + bj` where `j` is the imaginary unit. Rarely needed at beginner level.

````python
# Checking number types with type()

a = 42
b = 3.14
c = 3 + 4j

print(type(a))   # <class 'int'>
print(type(b))   # <class 'float'>
print(type(c))   # <class 'complex'>
````

````python
# Python integers have unlimited precision — no overflow!
big_number = 2 ** 100
print(big_number)
print(type(big_number))  # Still an int!
````

````python
# Type conversion between numeric types
x = 7          # int
y = float(x)   # Convert int to float
z = int(3.9)   # Convert float to int — truncates, does NOT round!

print(x, type(x))
print(y, type(y))
print(z, type(z))  # z = 3, not 4!
````

> **Key Insight:** `int(3.9)` gives `3`, not `4`. Python truncates (cuts off) the decimal, it does not round. To round, use `round(3.9)` which gives `4`.

## 3. Arithmetic Operators

Python provides 7 core arithmetic operators:

| Operator | Name | Example | Result |
|----------|------|---------|--------|
| `+` | Addition | `5 + 3` | `8` |
| `-` | Subtraction | `5 - 3` | `2` |
| `*` | Multiplication | `5 * 3` | `15` |
| `/` | Division (float) | `5 / 2` | `2.5` |
| `//` | Floor Division | `5 // 2` | `2` |
| `%` | Modulo (remainder) | `5 % 2` | `1` |
| `**` | Exponentiation | `5 ** 2` | `25` |

### Important Rule: Division Always Returns Float

In Python 3, the `/` operator **always** returns a float, even when dividing two integers evenly. This is intentional — Python 3 fixed a notorious Python 2 bug where `5 / 2` returned `2` instead of `2.5`.

````python
# Basic arithmetic — the building blocks

print(10 + 3)    # Addition:       13
print(10 - 3)    # Subtraction:    7
print(10 * 3)    # Multiplication: 30
print(10 / 3)    # Division:       3.3333... (always float)
print(10 / 2)    # Division:       5.0       (float, even though evenly divisible)
print(10 // 3)   # Floor Division: 3         (integer result)
print(10 % 3)    # Modulo:         1         (remainder)
print(10 ** 3)   # Exponentiation: 1000      (10 cubed)
````

````python
# Real-world example: Shopping cart
# You bought 4 items at $12.50 each and have a $10 discount coupon

item_price = 12.50
quantity = 4
discount = 10.0
tax_rate = 0.08  # 8% tax

subtotal = item_price * quantity
after_discount = subtotal - discount
tax_amount = after_discount * tax_rate
total = after_discount + tax_amount

print(f"Subtotal:      ${subtotal}")
print(f"After Discount: ${after_discount}")
print(f"Tax (8%):      ${tax_amount:.2f}")
print(f"Total:         ${total:.2f}")
````

````python
# Edge case: Operations with negative numbers

print(-7 + 3)    # -4   (negative + positive)
print(-7 - 3)    # -10  (negative - positive)
print(-7 * -3)   # 21   (negative * negative = positive)
print(-7 / 2)    # -3.5
print(-7 // 2)   # -4   (floor division floors toward negative infinity!)
print(-7 % 2)    #  1   (modulo with negatives — see detailed section below)
````

> **Critical Insight:** `-7 // 2` gives `-4`, not `-3`. Floor division always rounds **down** toward negative infinity, not toward zero. This is a common source of confusion.

````python
# Advanced example: Mixed int and float operations
# When you mix int and float, Python automatically promotes to float

result1 = 5 + 2.0    # int + float = float
result2 = 10 * 0.5   # int * float = float
result3 = 8 / 4      # int / int = float (always in Python 3)

print(result1, type(result1))   # 7.0 <class 'float'>
print(result2, type(result2))   # 5.0 <class 'float'>
print(result3, type(result3))   # 2.0 <class 'float'>
````

## 4. Integer Division (Floor Division) — `//`

### What Is It?

The `//` operator divides two numbers and **rounds the result down to the nearest whole number** (floors it). This is why it's called **floor division**.

### Why Does It Exist?

Many real-world problems require whole-number answers:
- How many **full boxes** can you pack with 47 items if each box holds 10? → `47 // 10 = 4`
- How many **complete weeks** are in 30 days? → `30 // 7 = 4`
- What **page number** is item 55 on, with 20 items per page? → `55 // 20 = 2` (page index 2)

### How It Works Internally

Floor division is equivalent to: `math.floor(a / b)`

The `floor` function always rounds toward **negative infinity** (not toward zero). This distinction matters for negative numbers.

````python
# Example 1: Basic floor division with positive numbers

print(17 // 5)    # 3  (17/5 = 3.4, floor = 3)
print(20 // 4)    # 5  (20/4 = 5.0, floor = 5)
print(1 // 3)     # 0  (1/3 = 0.33, floor = 0)
````

````python
# Example 2: Real-world — converting minutes to hours and minutes

total_minutes = 137

hours = total_minutes // 60    # How many complete hours?
remaining_minutes = total_minutes % 60  # Leftover minutes

print(f"{total_minutes} minutes = {hours} hour(s) and {remaining_minutes} minute(s)")
# Output: 137 minutes = 2 hour(s) and 17 minute(s)
````

````python
# Example 3: Floor division with floats — result is still a float!

print(7.5 // 2.0)   # 3.0 (result is float, but value is whole)
print(7.5 // 2)     # 3.0 (mixing int and float still gives float)
print(type(7 // 2)) # <class 'int'>   (int // int = int)
print(type(7.0 // 2)) # <class 'float'> (float // int = float)
````

````python
# Example 4: CRITICAL — Negative floor division
# Floor always goes toward NEGATIVE INFINITY, not toward zero

print(7 // 2)     #  3   (positive: rounds down from 3.5)
print(-7 // 2)    # -4   (negative: rounds DOWN from -3.5, which is -4, not -3!)
print(7 // -2)    # -4   (same reasoning)
print(-7 // -2)   #  3   (negative divided by negative = positive)

# To truncate toward zero instead, use int(a/b)
print(int(-7 / 2))  # -3  (truncates toward zero)
````

````python
# Example 5: Practical use — Pagination
# Given 85 items and 10 items per page, find total pages needed

total_items = 85
items_per_page = 10

full_pages = total_items // items_per_page       # 8 full pages
remaining = total_items % items_per_page         # 5 items on last page
total_pages = full_pages + (1 if remaining > 0 else 0)  # 9 total pages

print(f"Full pages: {full_pages}")
print(f"Items on last page: {remaining}")
print(f"Total pages needed: {total_pages}")
````

## 5. Modulo Operator — `%`

### What Is It?

The modulo operator (`%`) returns the **remainder** after division. It answers the question: *"After dividing as many whole times as possible, what is left over?"*

### Why Does It Exist?

Modulo is one of the most powerful and widely used operators in programming:
- **Checking if a number is even or odd** (`n % 2 == 0` means even)
- **Cycling through values** (days of the week, colors in a pattern)
- **Wrapping around** (game position resets after reaching boundary)
- **Checking divisibility** (`n % 5 == 0` means n is divisible by 5)

### Mathematical Relationship

The modulo operator satisfies this identity:
```
a = (a // b) * b + (a % b)
```
This means: `dividend = quotient * divisor + remainder`

In Python, the **sign of the result follows the sign of the divisor** (the right-hand number).

````python
# Example 1: Basic modulo

print(10 % 3)    # 1   (10 = 3*3 + 1)
print(15 % 5)    # 0   (15 = 5*3 + 0, perfectly divisible)
print(7 % 10)    # 7   (7 = 10*0 + 7, smaller than divisor)
print(1 % 2)     # 1   (1 is less than 2)
````

````python
# Example 2: Checking even/odd — the most common modulo use case

numbers = [1, 2, 7, 14, 23, 100, -4, -7]

for n in numbers:
    if n % 2 == 0:
        print(f"{n} is even")
    else:
        print(f"{n} is odd")
````

````python
# Example 3: Real-world — Day of the week calculator
# If today is Monday (day 0), what day will it be in 100 days?

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

current_day_index = 0   # Monday
days_ahead = 100

future_day_index = (current_day_index + days_ahead) % 7
print(f"In {days_ahead} days, it will be: {days[future_day_index]}")
````

````python
# Example 4: Edge case — modulo with negative numbers
# In Python, result sign matches the DIVISOR (right-hand side)

print(-7 % 3)    #  2   (result is positive because 3 is positive)
print(7 % -3)    # -2   (result is negative because -3 is negative)
print(-7 % -3)   # -1   (result is negative because -3 is negative)

# Verify with the identity: a = (a // b) * b + (a % b)
# -7 % 3: -7 = (-7 // 3) * 3 + (-7 % 3)
#         -7 = (-3) * 3 + 2
#         -7 = -9 + 2  = -7  CORRECT!
a, b = -7, 3
print(f"Verification: {a} = ({a // b}) * {b} + ({a % b}) = {(a // b) * b + (a % b)}")
````

````python
# Example 5: Advanced — Digital clock wrap-around
# If current hour is 22, what hour will it be in 5 hours?

current_hour = 22
hours_to_add = 5

new_hour = (current_hour + hours_to_add) % 24
print(f"Current hour: {current_hour}:00")
print(f"After {hours_to_add} hours: {new_hour}:00")
# Output: 3:00 (wraps around midnight)
````

## 6. Exponentiation — `**`

### What Is It?

The `**` operator raises a number to a **power** (exponent). `a ** b` means multiply `a` by itself `b` times.

### Why Does It Exist?

Exponentiation appears everywhere in computing:
- **Compound interest**: `principal * (1 + rate) ** years`
- **Data storage**: 1 GB = `2 ** 30` bytes
- **Scientific calculations**: Area of a circle = `pi * radius ** 2`
- **Machine learning**: Squaring errors in cost functions
- **Encryption**: RSA encryption is built on modular exponentiation

### How It Works Internally

Python uses an efficient algorithm called **fast exponentiation** (binary exponentiation), which computes `a ** b` in `O(log b)` steps rather than `O(b)` steps. That is why `2 ** 1000000` is fast even though it has a million multiplications conceptually.

````python
# Example 1: Basic exponentiation

print(2 ** 3)     # 8    (2 * 2 * 2)
print(5 ** 2)     # 25   (5 squared)
print(10 ** 6)    # 1000000 (one million)
print(2 ** 0)     # 1    (anything to the power of 0 is 1)
print(2 ** 1)     # 2    (anything to the power of 1 is itself)
````

````python
# Example 2: Fractional exponents = roots!
# Square root = ** 0.5, Cube root = ** (1/3)

print(25 ** 0.5)       # 5.0  (square root of 25)
print(27 ** (1/3))     # 3.0  (cube root of 27)
print(16 ** 0.25)      # 2.0  (4th root of 16)
print(64 ** (1/6))     # 2.0  (6th root of 64)
````

````python
# Example 3: Real-world — Compound interest formula
# A = P * (1 + r/n) ** (n*t)
# P = principal, r = annual rate, n = times compounded per year, t = years

principal = 10000      # $10,000 initial investment
annual_rate = 0.07     # 7% annual interest rate
times_per_year = 12    # Compounded monthly
years = 10             # Investment period

final_amount = principal * (1 + annual_rate / times_per_year) ** (times_per_year * years)
profit = final_amount - principal

print(f"Initial Investment: ${principal:,.2f}")
print(f"After {years} years:   ${final_amount:,.2f}")
print(f"Profit earned:     ${profit:,.2f}")
````

````python
# Example 4: Negative exponents — gives reciprocals (fractions)

print(2 ** -1)     # 0.5   (1/2)
print(2 ** -2)     # 0.25  (1/4)
print(10 ** -3)    # 0.001 (1/1000)
````

````python
# Example 5: Advanced — Computing storage sizes
# Computer memory uses powers of 2

print(f"1 KB = {2**10:,} bytes")
print(f"1 MB = {2**20:,} bytes")
print(f"1 GB = {2**30:,} bytes")
print(f"1 TB = {2**40:,} bytes")

# How many bytes in a 256 GB USB drive?
usb_size_gb = 256
bytes_in_usb = usb_size_gb * 2**30
print(f"\n256 GB USB = {bytes_in_usb:,} bytes")
````

## 7. Floating-Point Precision

### The Surprising Behavior

This is one of the most important concepts to understand as a programmer. Run this:

````python
# The famous floating-point surprise
print(0.1 + 0.2)          # You might expect 0.3...
print(0.1 + 0.2 == 0.3)   # Is it equal to 0.3?
````

### Why Does This Happen?

This is NOT a Python bug. It is a fundamental limitation of how computers store decimal numbers.

**The core problem:** Computers store numbers in binary (base 2). Just like `1/3` cannot be written exactly in decimal (it is `0.333...` forever), fractions like `0.1` and `0.2` **cannot be represented exactly in binary**.

- In binary, `0.1` is actually `0.00011001100110011...` (repeating forever)
- The computer stores only a fixed number of bits (64), so it rounds
- When you add two rounded numbers, the rounding errors accumulate

**This affects every programming language** — C, Java, JavaScript, Ruby — not just Python.

### How to Handle Floating-Point Issues

````python
# Solution 1: Use round() for display

result = 0.1 + 0.2
print(round(result, 2))          # 0.3  (rounds to 2 decimal places)
print(round(result, 10) == 0.3)  # True (rounding to 10 places is enough)
````

````python
# Solution 2: Use math.isclose() for comparisons — the RIGHT way to compare floats
import math

a = 0.1 + 0.2
b = 0.3

# WRONG: Direct comparison
print(a == b)                   # False (wrong approach)

# RIGHT: Tolerance-based comparison
print(math.isclose(a, b))       # True
print(math.isclose(a, b, rel_tol=1e-9))  # True with 9-decimal tolerance
````

````python
# Solution 3: Use the decimal module for financial calculations
# When exact decimal arithmetic is required (money, accounting)

from decimal import Decimal

# Float version (imprecise)
float_result = 0.1 + 0.2
print(f"Float:   {float_result}")

# Decimal version (exact)
decimal_result = Decimal('0.1') + Decimal('0.2')   # Use strings for exact input!
print(f"Decimal: {decimal_result}")
print(f"Decimal equals 0.3: {decimal_result == Decimal('0.3')}")
````

````python
# Understanding float precision limits

import sys
print(f"Float max value:       {sys.float_info.max:.2e}")
print(f"Float smallest value:  {sys.float_info.min:.2e}")
print(f"Float epsilon:         {sys.float_info.epsilon:.2e}")
# Epsilon is the smallest difference Python can detect between 1.0 and the next float
````

````python
# Practical floating-point formatting

pi_approx = 22 / 7
print(f"{pi_approx}")          # Full precision
print(f"{pi_approx:.2f}")      # 2 decimal places
print(f"{pi_approx:.5f}")      # 5 decimal places
print(f"{pi_approx:.10f}")     # 10 decimal places
print(round(pi_approx, 4))     # Rounds to 4 places
````

## 8. Operator Precedence

### What Is Operator Precedence?

When you write `3 + 5 * 2`, Python needs to decide: does it add first or multiply first? This ordering rule is called **operator precedence** (also called "order of operations").

Python follows the same rules you learned in school — **PEMDAS/BODMAS**:

| Priority | Operator(s) | Name |
|----------|-------------|------|
| 1 (Highest) | `()` | Parentheses |
| 2 | `**` | Exponentiation |
| 3 | `+x`, `-x` | Unary plus/minus |
| 4 | `*`, `/`, `//`, `%` | Multiplication, Division, Floor, Modulo |
| 5 (Lowest) | `+`, `-` | Addition, Subtraction |

**When operators have equal precedence**, Python evaluates **left to right** (except `**` which is right to left).

````python
# Example 1: Basic precedence

print(3 + 5 * 2)       # 13, not 16 (multiplication first)
print((3 + 5) * 2)     # 16 (parentheses override)

print(10 - 4 / 2)      # 8.0 (division first: 10 - 2.0 = 8.0)
print((10 - 4) / 2)    # 3.0 (parentheses first: 6 / 2 = 3.0)
````

````python
# Example 2: Exponentiation is right-to-left!
# This is different from most other operators

print(2 ** 3 ** 2)       # 512 (evaluated as 2 ** (3**2) = 2**9 = 512)
print((2 ** 3) ** 2)     # 64  (evaluated as (8) ** 2 = 64)

# Unary minus interacts carefully with exponentiation
print(-2 ** 2)           # -4  (-(2**2), NOT (-2)**2)
print((-2) ** 2)         # 4   ((-2) squared = 4)
````

````python
# Example 3: Complex expression breakdown
# Let's trace through: 2 + 3 * 4 ** 2 - 10 // 3

expression = 2 + 3 * 4 ** 2 - 10 // 3

step1 = 4 ** 2        # = 16  (exponentiation first)
step2 = 3 * step1     # = 48  (multiplication)
step3 = 10 // 3       # = 3   (floor division)
step4 = 2 + step2     # = 50  (left-to-right addition)
step5 = step4 - step3 # = 47  (subtraction)

print(f"4 ** 2 = {step1}")
print(f"3 * {step1} = {step2}")
print(f"10 // 3 = {step3}")
print(f"2 + {step2} = {step4}")
print(f"{step4} - {step3} = {step5}")
print(f"\nFull expression result: {expression}")
````

````python
# Example 4: Real-world — BMI Calculator
# BMI = weight(kg) / height(m) ** 2
# Precedence matters here!

weight_kg = 70
height_m = 1.75

# CORRECT: ** has higher precedence than /, so height is squared first
bmi_correct = weight_kg / height_m ** 2

# WRONG: This divides weight by height, then squares the result (not BMI!)
bmi_wrong = (weight_kg / height_m) ** 2

print(f"Correct BMI: {bmi_correct:.2f}")
print(f"Wrong result: {bmi_wrong:.2f}")
````

````python
# Example 5: When in doubt, use parentheses — it's a best practice!
# Parentheses make intent crystal clear

# Ambiguous (relies on precedence knowledge):
result1 = 100 - 5 * 2 ** 3 + 10 // 3

# Clear and explicit:
result2 = 100 - (5 * (2 ** 3)) + (10 // 3)

print(result1)   # Both give same answer
print(result2)   # But result2 is much easier to read and verify
````

> **Best Practice:** Even if you know the precedence rules, **use parentheses generously** to make your code readable. Your future self — and your teammates — will thank you.

## 9. The Math Module

### What Is It?

Python's built-in `math` module provides **mathematical functions and constants** beyond basic arithmetic. It is part of the Python Standard Library — no installation needed.

### Why Does It Exist?

Basic operators (`+`, `-`, `*`, `/`) are enough for simple tasks. But for:
- Trigonometry (sine, cosine)
- Logarithms and exponentials
- Rounding variations
- Special values like `pi` and `e`
- Square roots and powers

...you need the `math` module.

````python
import math

# --- Constants ---
print(f"pi  = {math.pi}")
print(f"e   = {math.e}")
print(f"tau = {math.tau}")   # tau = 2*pi
print(f"inf = {math.inf}")   # Positive infinity
print(f"nan = {math.nan}")   # Not a Number
````

````python
import math

# --- Rounding Functions ---
print("--- Rounding ---")
print(math.floor(3.7))    #  3  (always rounds DOWN, toward -infinity)
print(math.ceil(3.2))     #  4  (always rounds UP, toward +infinity)
print(math.trunc(3.7))    #  3  (truncates, toward zero)
print(math.trunc(-3.7))   # -3  (truncates toward zero, unlike floor)
print(round(3.5))         #  4  (Python's built-in rounds to nearest even)
print(round(4.5))         #  4  (banker's rounding: nearest even is 4)
````

````python
import math

# --- Power and Root Functions ---
print("--- Power & Root ---")
print(math.sqrt(144))         # 12.0  (square root)
print(math.pow(2, 10))        # 1024.0 (note: always returns float, unlike **)
print(math.isqrt(17))         # 4     (integer square root, Python 3.8+)
print(math.isqrt(16))         # 4     (exact integer square root)

# Cube root via ** (math module doesn't have cbrt in older versions)
print(27 ** (1/3))            # ~3.0  (floating point approach)
print(math.cbrt(27))          # 3.0   (available Python 3.11+)
````

````python
import math

# --- Logarithms ---
print("--- Logarithms ---")
print(math.log(math.e))      # 1.0   (natural log, base e)
print(math.log(100, 10))     # 2.0   (log base 10 of 100)
print(math.log10(1000))      # 3.0   (log base 10, more accurate)
print(math.log2(1024))       # 10.0  (log base 2)

# Real use: How many bits needed to store a number?
number = 500
bits_needed = math.ceil(math.log2(number))
print(f"\nBits needed to store {number}: {bits_needed}")
````

````python
import math

# --- Trigonometry (angles in radians!) ---
print("--- Trigonometry ---")
print(math.sin(0))                    # 0.0
print(math.cos(0))                    # 1.0
print(math.sin(math.pi / 2))         # 1.0  (sin of 90 degrees)
print(math.cos(math.pi))             # -1.0 (cos of 180 degrees)

# Convert degrees to radians and vice versa
print(math.radians(180))             # 3.14... (converts 180 degrees to radians)
print(math.degrees(math.pi))         # 180.0 (converts pi radians to degrees)

# Practical: Height of a tree
# If you stand 50m from a tree and the angle to the top is 35 degrees:
distance = 50  # meters
angle_degrees = 35
angle_radians = math.radians(angle_degrees)
tree_height = distance * math.tan(angle_radians)
print(f"\nTree height: {tree_height:.2f} meters")
````

````python
import math

# --- Other Useful Functions ---
print(math.factorial(5))     # 120  (5! = 5*4*3*2*1)
print(math.gcd(48, 36))      # 12   (Greatest Common Divisor)
print(math.lcm(4, 6))        # 12   (Least Common Multiple, Python 3.9+)
print(math.fabs(-7.5))       # 7.5  (absolute value, returns float)
print(abs(-7.5))             # 7.5  (built-in abs, works for all types)
print(math.hypot(3, 4))      # 5.0  (hypotenuse: sqrt(3^2 + 4^2))
````

## 10. Edge Cases & Common Mistakes

### Mistake 1: Division by Zero

````python
# Division by zero — always a ZeroDivisionError

try:
    print(10 / 0)
except ZeroDivisionError as e:
    print(f"Error: {e}")

try:
    print(10 // 0)
except ZeroDivisionError as e:
    print(f"Error: {e}")

# Best practice: Always check before dividing
def safe_divide(a, b):
    if b == 0:
        return None  # or return 0, or raise a custom error
    return a / b

print(safe_divide(10, 2))   # 5.0
print(safe_divide(10, 0))   # None
````

### Mistake 2: Comparing Floats Directly

````python
import math

# WRONG: Direct comparison
result = 0.1 + 0.2
print(result == 0.3)              # False! Dangerous in real code

# RIGHT: Use math.isclose()
print(math.isclose(result, 0.3))  # True

# For financial code, use Decimal module
from decimal import Decimal
print(Decimal('0.1') + Decimal('0.2') == Decimal('0.3'))  # True
````

### Mistake 3: Misunderstanding `-2 ** 2`

````python
# A very common mistake for beginners

print(-2 ** 2)       # -4 (this is -(2**2), not (-2)**2)
print((-2) ** 2)     # 4  (this is what most people intend)

# The reason: unary minus has LOWER precedence than **
````

### Mistake 4: int() truncates, does not round

````python
# Beginners often expect int() to round

print(int(3.9))     # 3 (truncated, not rounded!)
print(int(-3.9))    # -3 (truncated toward zero, not -4)

# If you want rounding, use round()
print(round(3.9))   # 4
print(round(3.5))   # 4 (banker's rounding: rounds to nearest even)
print(round(4.5))   # 4 (rounds to nearest even, not 5!)
````

### Mistake 5: Modulo with Negative Numbers

````python
# Python's modulo follows the sign of the DIVISOR
# This is mathematically consistent but surprises people from other languages

print(-7 % 3)    # 2  (positive, follows divisor)
print(7 % -3)    # -2 (negative, follows divisor)

# In C/Java/JavaScript: -7 % 3 would give -1 (sign follows dividend)
# Python's behavior is actually more mathematically correct

# To get C-style behavior in Python:
import math
print(math.fmod(-7, 3))   # -1.0 (sign follows dividend)
````

### Mistake 6: Augmented Assignment Operators

````python
# Python has shorthand augmented assignment operators
# These are very useful and commonly used

x = 10
x += 5     # Same as: x = x + 5
print(x)   # 15

x -= 3     # x = x - 3
print(x)   # 12

x *= 2     # x = x * 2
print(x)   # 24

x //= 5    # x = x // 5
print(x)   # 4

x **= 3    # x = x ** 3
print(x)   # 64

x %= 10    # x = x % 10
print(x)   # 4
````

## 11. Best Practices

### When to Use Each Operator

| Situation | Recommended Approach |
|-----------|---------------------|
| Need whole number result | Use `//` instead of `/` then casting to int |
| Checking divisibility | Use `% == 0` |
| Financial calculations | Use `decimal.Decimal` for exactness |
| Comparing floats | Use `math.isclose()` |
| Square root | Prefer `math.sqrt()` over `** 0.5` for clarity |
| Incrementing a counter | Use `+=` shorthand |
| Complex expressions | Add parentheses for readability |

### Code Style Tips

````python
import math

# Use spaces around operators (PEP 8 style)
result = 2 + 3 * 4    # GOOD
result = 2+3*4        # BAD (harder to read)

# Use parentheses for clarity in complex expressions
area = math.pi * (radius := 5) ** 2    # Compact but uses walrus operator
area = math.pi * (5 ** 2)              # Clearer

# Name magic numbers — don't use raw numbers in calculations
SECONDS_IN_HOUR = 3600
DAYS_IN_WEEK = 7

total_seconds = 2.5 * SECONDS_IN_HOUR    # GOOD: clear meaning
total_seconds = 2.5 * 3600               # BAD: 3600 is a magic number
````

## 12. Real-World Applications

### Application 1: Temperature Converter

````python
def celsius_to_fahrenheit(c):
    """Convert Celsius to Fahrenheit: F = (C * 9/5) + 32"""
    return (c * 9 / 5) + 32

def fahrenheit_to_celsius(f):
    """Convert Fahrenheit to Celsius: C = (F - 32) * 5/9"""
    return (f - 32) * 5 / 9

def celsius_to_kelvin(c):
    """Convert Celsius to Kelvin: K = C + 273.15"""
    return c + 273.15

temps_celsius = [0, 20, 37, 100, -40]

print(f"{'Celsius':>10} {'Fahrenheit':>12} {'Kelvin':>10}")
print("-" * 36)
for c in temps_celsius:
    f = celsius_to_fahrenheit(c)
    k = celsius_to_kelvin(c)
    print(f"{c:>10.1f} {f:>12.1f} {k:>10.2f}")
````

### Application 2: Grade Calculator with Statistics

````python
import math

# Student grades for 5 subjects
grades = [85, 92, 78, 96, 88]

# Calculate statistics using only math operations
total = sum(grades)
count = len(grades)
average = total / count

# Standard deviation (spread of grades)
variance = sum((g - average) ** 2 for g in grades) / count
std_dev = math.sqrt(variance)

highest = max(grades)
lowest = min(grades)
grade_range = highest - lowest

print(f"Grades: {grades}")
print(f"Total Score:        {total}")
print(f"Average:            {average:.1f}")
print(f"Highest:            {highest}")
print(f"Lowest:             {lowest}")
print(f"Range:              {grade_range}")
print(f"Standard Deviation: {std_dev:.2f}")
````

### Application 3: Time Duration Formatter

````python
def format_duration(total_seconds):
    """
    Convert a number of seconds into a human-readable string.
    Uses floor division and modulo to extract each time unit.
    """
    SECONDS_IN_MINUTE = 60
    SECONDS_IN_HOUR = 3600
    SECONDS_IN_DAY = 86400

    days    = total_seconds // SECONDS_IN_DAY
    remaining = total_seconds % SECONDS_IN_DAY

    hours   = remaining // SECONDS_IN_HOUR
    remaining = remaining % SECONDS_IN_HOUR

    minutes = remaining // SECONDS_IN_MINUTE
    seconds = remaining % SECONDS_IN_MINUTE

    parts = []
    if days:    parts.append(f"{days}d")
    if hours:   parts.append(f"{hours}h")
    if minutes: parts.append(f"{minutes}m")
    if seconds or not parts: parts.append(f"{seconds}s")

    return " ".join(parts)

# Test with various durations
test_cases = [45, 3600, 7384, 90000, 0, 172800]
for s in test_cases:
    print(f"{s:>8} seconds = {format_duration(s)}")
````

## 13. Summary

Here is a complete recap of everything covered in this lesson:

### Number Types
- `int`: Whole numbers, unlimited precision in Python
- `float`: Decimal numbers, stored in 64-bit IEEE 754 format
- Mixing `int` and `float` always produces a `float`
- `int(x)` truncates (does not round); `round(x)` rounds to nearest even

### Arithmetic Operators
- `+`, `-`, `*` work as expected
- `/` always returns `float` in Python 3
- `//` floor division: rounds down toward negative infinity
- `%` modulo: returns remainder; result sign follows divisor
- `**` exponentiation: evaluated right-to-left; supports fractional powers

### Floating-Point Precision
- `0.1 + 0.2 != 0.3` — this is not a Python bug, it is a binary representation limitation
- Use `math.isclose()` to compare floats
- Use `decimal.Decimal` for exact arithmetic (financial applications)

### Operator Precedence (PEMDAS order)
1. `()` Parentheses
2. `**` Exponentiation (right-to-left)
3. Unary `+x`, `-x`
4. `*`, `/`, `//`, `%`
5. `+`, `-`
- Use parentheses to make complex expressions readable

### Math Module Highlights
- Constants: `math.pi`, `math.e`
- Rounding: `math.floor()`, `math.ceil()`, `math.trunc()`
- Powers: `math.sqrt()`, `math.pow()`, `math.isqrt()`
- Logarithms: `math.log()`, `math.log10()`, `math.log2()`
- Trig: `math.sin()`, `math.cos()`, `math.tan()` (radians!)
- Utilities: `math.factorial()`, `math.gcd()`, `math.isclose()`, `math.hypot()`

````python
# Solution 2 — Leap Year Check

def is_leap_year(year):
    """A year is a leap year if divisible by 4,
    except century years must also be divisible by 400."""
    if year % 400 == 0:
        return True
    elif year % 100 == 0:
        return False
    elif year % 4 == 0:
        return True
    else:
        return False

test_years = [2024, 2023, 2000, 1900, 1996]
for y in test_years:
    status = "Leap year" if is_leap_year(y) else "Not a leap year"
    print(f"{y}: {status}")
````

````python
# Solution 4 — Temperature Converter

def convert_temperature(celsius):
    """Convert Celsius to Fahrenheit and Kelvin."""
    fahrenheit = (celsius * 9 / 5) + 32
    kelvin = celsius + 273.15
    return {
        'celsius': celsius,
        'fahrenheit': round(fahrenheit, 2),
        'kelvin': round(kelvin, 2)
    }

result = convert_temperature(25)
print(f"25°C = {result['fahrenheit']}°F = {result['kelvin']}K")
````

````python
# Solution 5 — Digit Sum Without String Conversion

def digit_sum(n):
    """Sum digits using modulo and floor division."""
    n = abs(n)   # Handle negative numbers
    total = 0
    while n > 0:
        total += n % 10    # Get last digit
        n = n // 10        # Remove last digit
    return total

print(digit_sum(12345))   # 1+2+3+4+5 = 15
print(digit_sum(9999))    # 9+9+9+9   = 36
print(digit_sum(-473))    # 4+7+3     = 14
````

````python
# Solution 6 — Circle Properties

import math

radius = 7

diameter = 2 * radius
circumference = 2 * math.pi * radius
area = math.pi * radius ** 2

print(f"Radius:        {radius}")
print(f"Diameter:      {diameter}")
print(f"Circumference: {circumference:.2f}")
print(f"Area:          {area:.2f}")
````

````python
# Solution 7 — Safe Square Root

import math

def safe_sqrt(n):
    """Returns sqrt(n) if n >= 0, else prints error message."""
    if n < 0:
        print(f"Cannot take sqrt of negative number: {n}")
        return None
    return math.sqrt(n)

print(safe_sqrt(25))    # 5.0
print(safe_sqrt(0))     # 0.0
print(safe_sqrt(-9))    # Error message
````

````python
# Solution 8 — Seconds to HH:MM:SS

def seconds_to_hhmmss(total_seconds):
    """Format total seconds as HH:MM:SS string."""
    hours   = total_seconds // 3600
    minutes = (total_seconds % 3600) // 60
    seconds = total_seconds % 60
    return f"{hours:02d}:{minutes:02d}:{seconds:02d}"

test_values = [0, 59, 3600, 9876, 86399]
for s in test_values:
    print(f"{s:>6} seconds = {seconds_to_hhmmss(s)}")
````

````python
# Solution 9 — Compound Interest

def compound_interest(principal, rate, times, years):
    """
    Calculate compound interest.
    principal: initial amount
    rate:      annual interest rate (as decimal, e.g. 0.06 for 6%)
    times:     number of times interest is compounded per year
    years:     investment duration in years
    """
    amount = principal * (1 + rate / times) ** (times * years)
    interest_earned = amount - principal
    return round(amount, 2), round(interest_earned, 2)

# $5000 at 6% compounded quarterly for 3 years
final, earned = compound_interest(5000, 0.06, 4, 3)
print(f"Final amount:    ${final:,.2f}")
print(f"Interest earned: ${earned:,.2f}")
````

````python
# Solution 10 — Number Palindrome (math-only approach)

def is_palindrome_number(n):
    """
    Check if an integer is a palindrome using only math.
    Strategy: Reverse the number mathematically, then compare.
    """
    if n < 0:
        return False  # Negative numbers are not palindromes

    original = n
    reversed_num = 0

    while n > 0:
        last_digit = n % 10        # Extract last digit
        reversed_num = reversed_num * 10 + last_digit  # Append to reversed
        n = n // 10                # Remove last digit

    return original == reversed_num

test_numbers = [121, 123, 1221, 12321, 10, 0, -121, 1001]
for num in test_numbers:
    result = "palindrome" if is_palindrome_number(num) else "not a palindrome"
    print(f"{num:>6} is {result}")
````

## 15. Mini Project: Personal Finance Calculator

### Project Description

Build a **Personal Finance Calculator** that performs the following tasks:

1. Calculate monthly salary from annual salary (and apply income tax)
2. Calculate monthly budget allocation using the 50/30/20 rule:
   - 50% for needs (rent, food, utilities)
   - 30% for wants (entertainment, dining out)
   - 20% for savings
3. Calculate how long it will take to save for a goal
4. Calculate total savings with compound interest

This project uses: arithmetic operators, division, modulo, exponentiation, `math` module, and float formatting.

````python
import math

# ============================================================
# PERSONAL FINANCE CALCULATOR
# ============================================================

def calculate_take_home(annual_salary, tax_rate):
    """Calculate monthly take-home pay after tax."""
    annual_tax = annual_salary * tax_rate
    annual_take_home = annual_salary - annual_tax
    monthly_take_home = annual_take_home / 12
    return monthly_take_home, annual_tax


def budget_50_30_20(monthly_income):
    """Split income using the 50/30/20 budgeting rule."""
    needs    = monthly_income * 0.50
    wants    = monthly_income * 0.30
    savings  = monthly_income * 0.20
    return needs, wants, savings


def months_to_reach_goal(monthly_savings, goal_amount):
    """Calculate how many months to save for a goal (no interest)."""
    if monthly_savings <= 0:
        return None
    months = math.ceil(goal_amount / monthly_savings)
    years  = months // 12
    extra_months = months % 12
    return months, years, extra_months


def savings_with_interest(monthly_savings, annual_rate, months):
    """
    Calculate future value of monthly savings with compound interest.
    FV = PMT * [((1 + r)^n - 1) / r]
    where r = monthly rate, n = number of months
    """
    monthly_rate = annual_rate / 12
    if monthly_rate == 0:
        return monthly_savings * months
    future_value = monthly_savings * (((1 + monthly_rate) ** months - 1) / monthly_rate)
    return future_value


# ============================================================
# RUN THE CALCULATOR
# ============================================================

ANNUAL_SALARY   = 60000    # $60,000 per year
TAX_RATE        = 0.25     # 25% income tax
SAVINGS_GOAL    = 20000    # Target: $20,000 emergency fund
INTEREST_RATE   = 0.05     # 5% annual savings rate

SEPARATOR = "=" * 50

# --- Income Analysis ---
monthly_income, annual_tax = calculate_take_home(ANNUAL_SALARY, TAX_RATE)

print(SEPARATOR)
print("INCOME SUMMARY")
print(SEPARATOR)
print(f"Annual Salary:      ${ANNUAL_SALARY:>10,.2f}")
print(f"Tax ({TAX_RATE*100:.0f}%):          ${annual_tax:>10,.2f}")
print(f"Annual Take-Home:   ${ANNUAL_SALARY - annual_tax:>10,.2f}")
print(f"Monthly Take-Home:  ${monthly_income:>10,.2f}")

# --- Budget Allocation ---
needs, wants, savings = budget_50_30_20(monthly_income)

print()
print(SEPARATOR)
print("MONTHLY BUDGET (50/30/20 Rule)")
print(SEPARATOR)
print(f"Needs (50%):        ${needs:>10,.2f}   (rent, food, bills)")
print(f"Wants (30%):        ${wants:>10,.2f}   (dining, entertainment)")
print(f"Savings (20%):      ${savings:>10,.2f}   (emergency fund, investing)")

# --- Goal Timeline ---
months, years, extra = months_to_reach_goal(savings, SAVINGS_GOAL)

print()
print(SEPARATOR)
print(f"SAVINGS GOAL: ${SAVINGS_GOAL:,.0f}")
print(SEPARATOR)
print(f"Monthly savings:    ${savings:>10,.2f}")
print(f"Without interest:   {months} months ({years}y {extra}m)")

# --- With Compound Interest ---
future_value = savings_with_interest(savings, INTEREST_RATE, months)
interest_earned = future_value - (savings * months)

print()
print(SEPARATOR)
print(f"WITH {INTEREST_RATE*100:.0f}% ANNUAL INTEREST (after {months} months)")
print(SEPARATOR)
print(f"Total Deposited:    ${savings * months:>10,.2f}")
print(f"Interest Earned:    ${interest_earned:>10,.2f}")
print(f"Final Balance:      ${future_value:>10,.2f}")
print(f"Goal reached early: Yes! {months} months vs {months} needed (with interest)")
print(SEPARATOR)
````
